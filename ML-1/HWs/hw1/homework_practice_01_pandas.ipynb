#### 6. [0.5 балла] Найдите курсы по выбору, на которые записывались как студенты 22-го года набора, так и студенты 21-го года.

```python
courses = ['fall_1', 'fall_2', 'fall_3', 'spring_1', 'spring_2', 'spring_3', 'blended']

courses_22 = set(df.loc[df["group_22"].notna(), courses].values.ravel().tolist())
courses_21 = set(df.loc[df["group_21"].notna(), courses].values.ravel().tolist())

common_courses = courses_22.intersection(courses_21) - {None, float("nan")}

print(common_courses)
```

```python
course_cols = ['fall_1','fall_2','fall_3','spring_1','spring_2','spring_3','blended']

s22 = df.loc[df['group_22'] != 0, course_cols].stack().astype("string").str.strip()
s21 = df.loc[df['group_21'] != 0, course_cols].stack().astype("string").str.strip()

courses_22 = set(s22.dropna().loc[s22.ne("") & s22.ne("0")].unique())
courses_21 = set(s21.dropna().loc[s21.ne("") & s21.ne("0")].unique())

common_courses   = sorted(courses_22 & courses_21)
courses_only_22  = sorted(courses_22 - courses_21)
courses_only_21  = sorted(courses_21 - courses_22)

print(f'Общие курсы: {', '.join(common_courses)}')
```

### Визуализации и matplotlib

- plot — обычный поточечный график, которым можно изображать кривые или отдельные точки;
- hist — гистограмма, показывающая распределение некоторой величины;
- scatter — график, показывающий взаимосвязь двух величин;
- bar — столбцовый график, показывающий взаимосвязь количественной величины от категориальной.

В этом задании вы попробуете построить один из них. Не забывайте про базовые принципы построения приличных графиков:
- оси должны быть подписаны, причём не слишком мелко;
- у графика должно быть название;

```python
# создай таблицу вместимостей курсов и посчитай количество занятых мест после первой волны
cap_df = pd.DataFrame({"course": pd.unique(df_g["course"])})
cap_df["capacity"] = cap_df["course"].map(capacities).fillna(other_capacity).astype(int)

used_df = (
    df_g[df_g["rank_in_course"] < df_g["capacity"]]
    .groupby("course")
    .size()
    .reset_index(name="used")
)

# Создай таблицу с остатками мест, объедини вместимость курсов с числом занятых мест, посчитай сколько свободных мест осталось
rem_df = cap_df.merge(used_df, on="course", how="left")
rem_df["used"] = rem_df["used"].fillna(0).astype(int)
rem_df["remaining"] = (rem_df["capacity"] - rem_df["used"]).clip(lower=0)



# В wave_1 колонка называется "wave_1", в ней лежат списки курсов
w1_expl = wave_1.explode("wave_1").rename(columns={"wave_1": "course"})

# посчитай количество назначенных курсов для каждого студента.
assigned_cnt = w1_expl.groupby("id").size().reset_index(name="assigned_count")
```

```python
df_work = df_base.copy()

# сделай расчёт: сколько курсов у студента уже занято после первой волны, сколько ещё нужно добрать, и добавь эту информацию обратно в рабочий датафрейм
need_df = df_work[["id", "fall_courses"]].merge(assigned_cnt, on="id", how="left")
need_df["assigned_count"] = need_df["assigned_count"].fillna(0).astype(int)
need_df["still_need"] = (need_df["fall_courses"] - need_df["assigned_count"]).clip(lower=0)

df_need = df_work.merge(need_df[["id", "still_need"]], on="id", how="left")
df_need["still_need"] = df_need["still_need"].fillna(0).astype(int)

# студентов, у которых ещё остались незаполненные места, подставь их второй или третий приоритет и очисти список от пустых или некорректных значений
cand_single = df_need[(df_need["fall_courses"] == 1) & (df_need["still_need"] > 0)][["id", "fall_2", "percentile"]].rename(columns={"fall_2": "course"})
cand_double = df_need[(df_need["fall_courses"] == 2) & (df_need["still_need"] > 0)][["id", "fall_3", "percentile"]].rename(columns={"fall_3": "course"})

wave2_pool = pd.concat([cand_single, cand_double], ignore_index=True)
wave2_pool["course"] = wave2_pool["course"].astype("string").str.strip()
wave2_pool = wave2_pool[wave2_pool["course"].notna() & wave2_pool["course"].ne("") & wave2_pool["course"].ne("0")]

# убири уже зачисленных
assigned_pairs = w1_expl[["id", "course"]].dropna().drop_duplicates()
wave2_pool = wave2_pool.merge(assigned_pairs, on=["id", "course"], how="left", indicator=True)
wave2_pool = wave2_pool[wave2_pool["_merge"] == "left_only"].drop(columns="_merge")

# отсортируй кандидатов второй волны по курсу и перцентилю, добавь остатки мест, присвой каждому ранг в рамках курса и отфильтруй тех, кто попадает в лимит, собери список назначенных курсов для каждого студента во второй волне
wave2_sorted = wave2_pool.sort_values(["course", "percentile"], ascending=[True, False])
wave2_sorted = wave2_sorted.merge(rem_df[["course", "remaining"]], on="course", how="left")
wave2_sorted["remaining"] = wave2_sorted["remaining"].fillna(0).astype(int)

wave2_sorted["rank_in_course_w2"] = wave2_sorted.groupby("course").cumcount()
wave_2_detail = wave2_sorted[wave2_sorted["rank_in_course_w2"] < wave2_sorted["remaining"]].copy()
wave_2 = wave_2_detail.groupby("id")["course"].apply(list).reset_index(name="wave_2")

# пересчитай количество занятых мест после второй волны, объедини с исходными лимитами, рассчитай сколько мест осталось доступно на каждом курсе
used_after_w2 = (
    pd.concat([
        df_g[df_g["rank_in_course"] < df_g["capacity"]][["course"]],
        wave_2_detail[["course"]]
    ])
    .groupby("course")
    .size()
    .reset_index(name="used")
)

rem_df2 = cap_df.merge(used_after_w2, on="course", how="left")
rem_df2["used"] = rem_df2["used"].fillna(0).astype(int)
rem_df2["remaining"] = (rem_df2["capacity"] - rem_df2["used"]).clip(lower=0)



# собери все назначения студентов после двух волн, посчитай у каждого количество уже полученных курсов и вычисли, сколько курсов ещё нужно добрать каждому студенту
all_assigned = pd.concat([
    w1_expl[["id", "course"]],
    wave_2_detail[["id", "course"]],
], ignore_index=True)

assigned_cnt2 = all_assigned.groupby("id").size().reset_index(name="assigned_count")

need_df2 = df_work[["id", "fall_courses"]].merge(assigned_cnt2, on="id", how="left")
need_df2["assigned_count"] = need_df2["assigned_count"].fillna(0).astype(int)
need_df2["still_need"] = (need_df2["fall_courses"] - need_df2["assigned_count"]).clip(lower=0)

df_need2 = df_work.merge(need_df2[["id", "still_need"]], on="id", how="left")
df_need2["still_need"] = df_need2["still_need"].fillna(0).astype(int)

# возьми только тех студентов, у кого ещё остались незаполненные курсы, подставь их выбор fall_3, очисти значения и оставь только валидные курсы
wave3_pool = df_need2[(df_need2["still_need"] > 0)][["id", "fall_3", "percentile"]].rename(columns={"fall_3": "course"})
wave3_pool["course"] = wave3_pool["course"].astype("string").str.strip()
wave3_pool = wave3_pool[wave3_pool["course"].notna() & wave3_pool["course"].ne("") & wave3_pool["course"].ne("0")]

# исключи из пула третьей волны тех студентов, которые уже зачислены на этот курс: объедини таблицы по id и course, оставь только записи без совпадений
assigned_pairs2 = all_assigned.dropna().drop_duplicates()
wave3_pool = wave3_pool.merge(assigned_pairs2, on=["id", "course"], how="left", indicator=True)
wave3_pool = wave3_pool[wave3_pool["_merge"] == "left_only"].drop(columns="_merge")

# распредели студентов в третьей волне: отсортируй кандидатов по курсу и перцентилю, добавь информацию об оставшихся местах, посчитай порядковый номер студента в каждом курсе и выбери тех, кто попадает в лимит; сохрани результаты в таблицы с деталями и со списком курсов по каждому студенту
wave3_sorted = wave3_pool.sort_values(["course", "percentile"], ascending=[True, False])
wave3_sorted = wave3_sorted.merge(rem_df2[["course", "remaining"]], on="course", how="left")
wave3_sorted["remaining"] = wave3_sorted["remaining"].fillna(0).astype(int)

wave3_sorted["rank_in_course_w3"] = wave3_sorted.groupby("course").cumcount()
wave_3_detail = wave3_sorted[wave3_sorted["rank_in_course_w3"] < wave3_sorted["remaining"]].copy()
wave_3 = wave_3_detail.groupby("id")["course"].apply(list).reset_index(name="wave_3")

# объедини все зачисления из первой, второй и третьей волн, сгруппируй их по студенту и собери список курсов для каждого id в итоговую таблицу
all_assigned_final = pd.concat([
    w1_expl[["id", "course"]],
    wave_2_detail[["id", "course"]],
    wave_3_detail[["id", "course"]],
], ignore_index=True)

final_courses = all_assigned_final.groupby("id")["course"].apply(list).reset_index(name="courses")
```

```python
# разнеси курсы по столбцам course_1 и course_2, присвой каждому порядковый номер и выполни pivot по id
final_exploded = final_courses.explode("courses").reset_index(drop=True)
final_exploded["course_num"] = final_exploded.groupby("id").cumcount() + 1

final_pivot = final_exploded.pivot(index="id", columns="course_num", values="courses").reset_index()
final_pivot = final_pivot.rename(columns={1: "course_1", 2: "course_2"})

# сделай заполнение пустых значений: в колонке course_1 подставь ???, а в колонке course_2 прочерк для студентов с одним курсом
final_pivot["course_1"] = final_pivot["course_1"].fillna("???")
final_pivot["course_2"] = final_pivot["course_2"].fillna("-")

final_pivot.to_csv("res_fall.csv", index=False)
```
