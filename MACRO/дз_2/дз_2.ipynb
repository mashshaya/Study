```python
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

Зависимая переменная и регрессоры

```python
y = df["Fiscal Balance"]
X = df[[
    "Fiscal Rule",
    "Oil Prices",
    "Oil Production",
    "GDP Annual Growth Rate",
    "Exchange Rates (USD)"
]]
X = sm.add_constant(X)
```

OLS-оценка модели

```python
model = sm.OLS(y, X).fit()
model_summary = model.summary()
model_summary
```

- F-тест: p < 0.0000001: регрессия в целом статистически значима.
- R_2 = 0.854: модель объясняет приемрно 85% вариации фискального баланса.

```python
model.params
```

Прогнозный набор данных

```python
future_years = [2026, 2027, 2028, 2029, 2030]

X_future = pd.DataFrame({
    "const": 1,
    "Fiscal Rule": [1, 1, 1, 1, 1],
    "Oil Prices": df["Oil Prices"].tail(5).values,
    "Oil Production": df["Oil Production"].tail(5).values,
    "GDP Annual Growth Rate": df["GDP Annual Growth Rate"].tail(5).values,
    "Exchange Rates (USD)": df["Exchange Rates (USD)"].tail(5).values
}, index=future_years)
 
```

```python
forecast = model.predict(X_future)
```

График факта и прогноза

```python
plt.figure(figsize=(10,5))
plt.plot(df["Date"], y, label="Факт")
plt.plot(future_years, forecast, label="Прогноз")
plt.xlabel("Год")
plt.ylabel("Fiscal Balance")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA

y = df["Fiscal Balance"]

model = ARIMA(y, order=(1,1,1)).fit()
forecast = model.forecast(steps=5)

print(forecast)

plt.plot(df["Date"], y, label="Факт")
plt.plot([2026,2027,2028,2029,2030], forecast, label="ARIMA прогноз")
plt.legend()
plt.show()
```
