## Упражнение 1.1

**Задание.** Используя уравнение Фоккера-Планка, доказать, что динамика Винеровского процесса не имеет стационарного распределения. Показать, что с точностью до константы единственная стационарная мера - мера Лебега.

**Решение.**

Для одномерного Винеровского процесса
$$
dX_t=dW_t
$$
генератор равен
$$
Af(x)=\frac12 f''(x).
$$
Он самосопряжён, поэтому уравнение Фоккера-Планка для плотности $p(t,x)$ имеет вид
$$
\partial_t p(t,x)=A^*p(t,x)=\frac12\partial_{xx}p(t,x).
$$

Стационарная плотность $p_*(x)$ должна удовлетворять
$$
0=\frac12p_*''(x).
$$
Значит,
$$
p_*(x)=ax+b.
$$
Если это неотрицательная плотность вероятности на всей прямой, то линейная функция должна быть неотрицательной на всей $\mathbb R$. Поэтому $a=0$ и $b\ge0$. Но при $b>0$
$$
\int_{\mathbb R}b\,dx=\infty,
$$
а при $b=0$ получаем нулевую меру, а не вероятность. Следовательно, у Винеровского процесса нет стационарного вероятностного распределения.

Если искать не вероятность, а стационарную локально конечную меру с плотностью, то остаётся
$$
\nu(dx)=c\,dx,\qquad c>0.
$$
Это мера Лебега с точностью до множителя:
$$
\nu((a,b))=c(b-a).
$$

Строгая версия той же идеи без предположения о гладкой плотности: если бы вероятность $\pi$ была инвариантной, то $\pi*N(0,t)=\pi$. По характеристическим функциям
$$
\widehat\pi(\xi)e^{-t\xi^2/2}=\widehat\pi(\xi).
$$
Для $\xi\ne0$ отсюда $\widehat\pi(\xi)=0$, но $\widehat\pi(0)=1$, что противоречит непрерывности характеристической функции в нуле.

## Упражнение 1.2

**Задание.** Функция $u(t,x_1,x_2)$ удовлетворяет
$$
\partial_tu+(\theta-x_1)\partial_{x_1}u+(\theta-x_2)\partial_{x_2}u
-2u-2\sigma^2(\partial_{x_1}u+\partial_{x_2}u)
=\sigma^2(x_1\partial^2_{x_1}u+x_2\partial^2_{x_2}u),
$$
с начальным условием $u(0,x_1,x_2)=f(x_1,x_2)$, где $f$ - плотность вероятности. Нужно подобрать двумерную однородную диффузию Ито с $X_0\sim f$ и $X_t\sim u(t,\cdot)$.

**Решение.**

Перенесём всё, кроме $\partial_tu$, вправо:
$$
\partial_tu
=(x_1-\theta)\partial_{x_1}u+(x_2-\theta)\partial_{x_2}u
+2u+2\sigma^2(\partial_{x_1}u+\partial_{x_2}u)
+\sigma^2(x_1\partial^2_{x_1}u+x_2\partial^2_{x_2}u).
$$

По лекционной формуле для прямого уравнения Колмогорова
$$
\partial_tu=A^*u,\qquad
A^*u=-\sum_i\partial_{x_i}(b_i u)
+\frac12\sum_{i,j}\partial_{x_ix_j}^2(a_{ij}u),
$$
где $a=\Sigma\Sigma^\top$. Подберём независимые координаты:
$$
\boxed{
\begin{aligned}
dX_t^1&=(\theta-X_t^1)\,dt+\sqrt{2\sigma^2X_t^1}\,dW_t^1,\\
dX_t^2&=(\theta-X_t^2)\,dt+\sqrt{2\sigma^2X_t^2}\,dW_t^2,
\end{aligned}}
$$
где $W^1,W^2$ - независимые Винеровские процессы, а $X_0\sim f$.

Для этой диффузии
$$
b_i(x)=\theta-x_i,\qquad a_{ii}(x)=2\sigma^2x_i,\qquad a_{12}=a_{21}=0.
$$
Раскрываем одно координатное слагаемое:
$$
-\partial_{x_i}\big((\theta-x_i)u\big)
=(x_i-\theta)\partial_{x_i}u+u,
$$
$$
\frac12\partial^2_{x_i}\big(2\sigma^2x_i u\big)
=\sigma^2(2\partial_{x_i}u+x_i\partial^2_{x_i}u).
$$
После суммирования по $i=1,2$ получаем ровно PDE из условия.

Технически важная оговорка: коэффициент $\sqrt{2\sigma^2x_i}$ вещественный только при $x_i\ge0$. Поэтому корректная область для этой модели - $\mathbb R_+^2$, начальная плотность $f$ должна быть сосредоточена на $\mathbb R_+^2$, а на границе нужны стандартные условия для CIR-типа процесса. Если понимать условие буквально на всей $\mathbb R^2$, то такой вещественной диффузии Ито с матрицей ковариации $2\sigma^2\operatorname{diag}(x_1,x_2)$ не существует: при отрицательных $x_i$ матрица перестаёт быть неотрицательно определённой.

## Упражнение 1.3

**Задание.** Пусть
$$
C(t,x)=\mathbb E\left[e^{-\mu(T-t)}(X_T-K)_+\mid X_t=x\right],
$$
где
$$
dX_t=\mu X_t\,dt+\sigma X_t\,dW_t.
$$
Вывести PDE для цены опциона и граничное условие.

**Решение.**

Берём обратное уравнение Колмогорова, потому что $C(t,x)$ - условное ожидание будущей выплаты при фиксированном текущем состоянии $X_t=x$. Генератор GBM:
$$
Lf(x)=\mu x f'(x)+\frac12\sigma^2x^2f''(x).
$$
Пусть $g(x)=(x-K)_+$. Формула Фейнмана-Каца даёт
$$
C(t,x)=\mathbb E_{t,x}\left[e^{-\mu(T-t)}g(X_T)\right]
$$
как решение задачи
$$
\boxed{
\partial_tC+\mu x\partial_xC+\frac12\sigma^2x^2\partial^2_{xx}C-\mu C=0,\qquad 0\le t<T,\ x>0.
}
$$
Терминальное условие:
$$
\boxed{C(T,x)=(x-K)_+.}
$$
Естественные граничные условия:
$$
C(t,0)=0,
$$
потому что $0$ - поглощающее состояние для GBM, и при $x\to\infty$
$$
C(t,x)\sim x-Ke^{-\mu(T-t)}.
$$

**Почему это действительно Фейнман-Кац.** Пусть $C$ достаточно гладкая и удовлетворяет PDE. Рассмотрим
$$
Y_s=e^{-\mu(s-t)}C(s,X_s),\qquad s\in[t,T].
$$
По формуле Ито
$$
dY_s=e^{-\mu(s-t)}
\left(\partial_sC+LC-\mu C\right)(s,X_s)\,ds
+e^{-\mu(s-t)}\sigma X_s\partial_xC(s,X_s)\,dW_s.
$$
По PDE дрейф равен нулю. Значит, $Y_s$ - локальный мартингал; для payoff call и GBM стандартные условия роста переводят его в настоящий мартингал на конечном горизонте. Поэтому
$$
C(t,x)=Y_t=\mathbb E[Y_T\mid X_t=x]
=\mathbb E\left[e^{-\mu(T-t)}(X_T-K)_+\mid X_t=x\right].
$$
Это и обосновывает выбор обратного, а не прямого уравнения: мы эволюционируем функцию цены назад от терминального условия.

## Упражнение 1.4

**Задание.** Пусть
$$
\tau_w=\inf\{t\ge0:W_t\ge w\},\qquad w\ge0.
$$
Доказать, что $\tau_w$ - момент остановки, что $(\tau_w)_{w\ge0}$ - процесс Леви, и найти плотность сечения.

**Решение.**

Для фиксированного $w$:
$$
\{\tau_w\le t\}=\left\{\sup_{0\le s\le t}W_s\ge w\right\}.
$$
Так как $W$ адаптирован и имеет непрерывные траектории, $\sup_{0\le s\le t}W_s$ является $\mathcal F_t^W$-измеримой случайной величиной. Значит,
$$
\{\tau_w\le t\}\in\mathcal F_t^W,
$$
то есть $\tau_w$ - момент остановки.

Теперь рассматриваем $w\mapsto\tau_w$ как процесс по уровню. Очевидно, $\tau_0=0$ и траектории неубывают. Для $v,w\ge0$
$$
\tau_{w+v}-\tau_w
$$
есть время, за которое после достижения уровня $w$ процесс впервые поднимется ещё на $v$. По сильному марковскому свойству
$$
\widetilde W_t=W_{\tau_w+t}-W_{\tau_w}
$$
является новым Винеровским процессом, независимым от прошлого до $\tau_w$. Поэтому
$$
\tau_{w+v}-\tau_w\stackrel{d}{=}\tau_v
$$
и это приращение независимо от $\sigma(\tau_u:u\le w)$. Аналогично получаем независимость всех непересекающихся приращений. Стационарность приращений следует из зависимости только от разности уровней.

Наконец, стохастическая непрерывность в нуле следует из формулы отражения:
$$
\mathbb P(\tau_h>\varepsilon)
=1-2\left(1-\Phi\left(\frac{h}{\sqrt\varepsilon}\right)\right)\to0,\qquad h\downarrow0.
$$
Значит, $(\tau_w)_{w\ge0}$ - процесс Леви; точнее, возрастающий процесс Леви, то есть субординатор.

По принципу отражения:
$$
\mathbb P(\tau_w\le t)=
\mathbb P\left(\sup_{0\le s\le t}W_s\ge w\right)
=2\mathbb P(W_t\ge w)
=2\left(1-\Phi\left(\frac{w}{\sqrt t}\right)\right).
$$
Дифференцируя по $t$, получаем плотность:
$$
\boxed{
f_{\tau_w}(t)=\frac{w}{\sqrt{2\pi t^3}}\exp\left(-\frac{w^2}{2t}\right),
\qquad t>0,\ w>0.
}
$$
Также
$$
\mathbb E e^{-q\tau_w}=e^{-w\sqrt{2q}}.
$$

Терминологическая аккуратность: для стандартного Винеровского процесса без сноса это $1/2$-устойчивый субординатор с распределением Леви для каждого сечения. Inverse Gaussian получается для времени достижения броуновского движения со сносом, а Normal Inverse Gaussian обычно строится через подчинение Brownian motion inverse-Gaussian субординатором. Поэтому слово `NIG` в условии лучше понимать как ссылку на этот родственный класс; в нулевом сносе плотность выше - предельный случай.

## Упражнение 1.5

**Задание.** Для GBM
$$
X_t=X_0e^{(\mu-\sigma^2/2)t+\sigma W_t}
$$
при $\mu<\sigma^2/2$ показать, что $X_t\to0$ почти наверное, и найти вероятность превзойти уровень $R$ при $X_0=x<R$.

**Решение.**

Положим
$$
a=\mu-\frac{\sigma^2}{2}<0.
$$
Тогда
$$
\log X_t=\log X_0+at+\sigma W_t.
$$
По усиленному закону больших чисел для Винеровского процесса
$$
\frac{W_t}{t}\to0\quad\text{п.н.}
$$
Следовательно,
$$
\frac{\log X_t}{t}\to a<0,
$$
то есть $\log X_t\to-\infty$, и
$$
\boxed{X_t\to0\quad\text{п.н.}}
$$

Теперь найдём
$$
\mathbb P_x\left(\sup_{t\ge0}X_t\ge R\right),\qquad x<R.
$$
Переход к логарифму даёт
$$
\mathbb P\left(\sup_{t\ge0}(\sigma W_t+at)\ge m\right),
\qquad m=\log\frac{R}{x}>0.
$$
Обозначим $Y_t=\sigma W_t+at$. Его генератор:
$$
Lf(y)=a f'(y)+\frac{\sigma^2}{2}f''(y).
$$
Пусть
$$
q(y)=\mathbb P_y(\tau_m<\infty),\qquad \tau_m=\inf\{t:Y_t\ge m\}.
$$
По формуле Дынкина/задаче Дирихле $Lq=0$ на $(-\infty,m)$ с условиями $q(m)=1$ и $q(-\infty)=0$. Решение имеет вид
$$
q(y)=A+B e^{-2ay/\sigma^2}.
$$
Так как $a<0$, при $y\to-\infty$ экспонента стремится к $0$, значит $A=0$. Из $q(m)=1$ получаем
$$
q(0)=e^{2am/\sigma^2}.
$$
Значит,
$$
\boxed{
\mathbb P_x\left(\sup_{t\ge0}X_t\ge R\right)
=\exp\left(\frac{2(\mu-\sigma^2/2)}{\sigma^2}\log\frac{R}{x}\right)
=\left(\frac{x}{R}\right)^{1-2\mu/\sigma^2}.
}
$$
Так как $\mu<\sigma^2/2$, показатель $1-2\mu/\sigma^2$ положителен, и вероятность лежит в $(0,1)$.

## Упражнение 1.6

**Задание.** Для GBM
$$
X_t=X_0e^{(\mu-\sigma^2/2)t+\sigma W_t}
$$
при $\mu>\sigma^2/2$ показать, что $X_t\to\infty$ почти наверное, и найти вероятность упасть ниже уровня $R$ при $X_0=x>R$.

**Решение.**

Снова положим
$$
a=\mu-\frac{\sigma^2}{2}>0.
$$
Тогда
$$
\log X_t=\log X_0+at+\sigma W_t.
$$
Поскольку
$$
\frac{W_t}{t}\to0\quad\text{п.н.},
$$
получаем
$$
\frac{\log X_t}{t}\to a>0.
$$
Значит $\log X_t\to+\infty$, и
$$
\boxed{X_t\to\infty\quad\text{п.н.}}
$$

Нужно найти
$$
\mathbb P_x\left(\inf_{t\ge0}X_t\le R\right),
\qquad x>R.
$$
Это эквивалентно
$$
\mathbb P\left(\inf_{t\ge0}(\sigma W_t+at)\le -m\right),
\qquad m=\log\frac{x}{R}>0.
$$
Для $Y_t=\sigma W_t+at$ рассмотрим
$$
r(y)=\mathbb P_y(\tau_{-m}<\infty),\qquad
\tau_{-m}=\inf\{t:Y_t\le -m\}.
$$
Снова решаем $Lr=0$:
$$
a r'(y)+\frac{\sigma^2}{2}r''(y)=0,\qquad y>-m,
$$
с условиями $r(-m)=1$ и $r(+\infty)=0$. Общее решение
$$
r(y)=A+B e^{-2ay/\sigma^2}.
$$
Так как $a>0$, при $y\to+\infty$ экспонента стремится к $0$, значит $A=0$. Из $r(-m)=1$:
$$
B e^{2am/\sigma^2}=1,
$$
поэтому
$$
r(0)=e^{-2am/\sigma^2}.
$$
Следовательно,
$$
\boxed{
\mathbb P_x\left(\inf_{t\ge0}X_t\le R\right)
=\exp\left(-\frac{2(\mu-\sigma^2/2)}{\sigma^2}\log\frac{x}{R}\right)
=\left(\frac{R}{x}\right)^{2\mu/\sigma^2-1}.
}
$$
Так как $\mu>\sigma^2/2$, показатель $2\mu/\sigma^2-1$ положителен, и вероятность лежит в $(0,1)$.
