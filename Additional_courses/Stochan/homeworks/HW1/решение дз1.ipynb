## Листок 1. Случайные процессы: определения

### Упражнение 1.1

Пусть $U\sim \operatorname{Bernoulli}(1/3)$ и $V\sim U[0,1]$ независимы, а
$$
X_t=(Ut^2+1)e^{-tV}.
$$

Нужно описать траектории, конечномерные распределения, матожидание и ковариационную функцию.

**Решение.**

Если зафиксировать исход $\omega$, то фиксируются два числа: $U(\omega)\in\{0,1\}$ и $V(\omega)=v\in[0,1]$. Поэтому траектории имеют один из двух видов:
$$
X_t=e^{-tv},\quad \text{если } U=0,
$$
или
$$
X_t=(t^2+1)e^{-tv},\quad \text{если } U=1.
$$
Первый вариант возникает с вероятностью $2/3$, второй - с вероятностью $1/3$.

Для любого конечного набора моментов $t_1,\ldots,t_n$ распределение вектора
$(X_{t_1},\ldots,X_{t_n})$ задается смесью:
$$
\mathbb P\{(X_{t_1},\ldots,X_{t_n})\in A\}
=\frac23\int_0^1 \mathbf 1_A(e^{-t_1v},\ldots,e^{-t_nv})\,dv
+\frac13\int_0^1 \mathbf 1_A((1+t_1^2)e^{-t_1v},\ldots,(1+t_n^2)e^{-t_nv})\,dv.
$$

Матожидание:
$$
m(t)=\mathbb E X_t
=\left(1+\frac{t^2}{3}\right)\int_0^1 e^{-tv}\,dv
=\left(1+\frac{t^2}{3}\right)\frac{1-e^{-t}}{t},
$$
при $t=0$ понимаем по пределу: $m(0)=1$.

Для второго момента:
$$
\mathbb E[X_tX_s]
=\mathbb E[(Ut^2+1)(Us^2+1)]\mathbb E[e^{-(t+s)V}].
$$
Так как $U^2=U$ и $\mathbb E U=1/3$,
$$
\mathbb E[(Ut^2+1)(Us^2+1)]
=1+\frac{t^2+s^2+t^2s^2}{3}.
$$
Значит,
$$
\mathbb E[X_tX_s]
=\left(1+\frac{t^2+s^2+t^2s^2}{3}\right)\frac{1-e^{-(t+s)}}{t+s},
$$
где при $t+s=0$ снова берется предел. Ковариационная функция:
$$
K(t,s)=\left(1+\frac{t^2+s^2+t^2s^2}{3}\right)\frac{1-e^{-(t+s)}}{t+s}-m(t)m(s).
$$


### Упражнение 1.2

Пусть $(\varepsilon_t)_{t\in\mathbb Z}$ - гауссовский белый шум с $\mathbb E\varepsilon_t=0$ и $\operatorname{Var}\varepsilon_t=1$.
Процесс задан формулой
$$
X_{t+1}=\alpha\varepsilon_t+\beta\varepsilon_{t-1}+\gamma\varepsilon_{t-2}+\varepsilon_{t+1}.
$$

**Решение.**

Удобно переписать процесс как
$$
X_t=\varepsilon_t+\alpha\varepsilon_{t-1}+\beta\varepsilon_{t-2}+\gamma\varepsilon_{t-3}.
$$
Тогда
$$
\mathbb E X_t=0.
$$

Ковариация зависит только от лага $h$:
$$
\operatorname{Cov}(X_t,X_{t+h})=
\begin{cases}
1+\alpha^2+\beta^2+\gamma^2, & h=0,\\
\alpha+\alpha\beta+\beta\gamma, & |h|=1,\\
\beta+\alpha\gamma, & |h|=2,\\
\gamma, & |h|=3,\\
0, & |h|>3.
\end{cases}
$$
Следовательно, процесс стационарен в широком смысле при любых $\alpha,\beta,\gamma\in\mathbb R$: матожидание постоянно, а ковариация зависит только от разности времен.


### Упражнение 1.3

Пусть $X$ - гауссовский процесс с нулевым матожиданием. Нужно проверить стационарность в широком смысле для двух ковариационных функций.

**Решение.**

1. Для
$$
K(t,s)=(\sin t\sin s)^2
$$
дисперсия равна
$$
K(t,t)=\sin^4 t,
$$
она зависит от $t$. Поэтому процесс не стационарен в широком смысле.

2. Для функции Орнштейна-Уленбека
$$
K(t,s)=e^{-|t-s|/(2l)},\quad l>0,
$$
дисперсия постоянна: $K(t,t)=1$, а ковариация зависит только от $|t-s|$. Поэтому процесс стационарен в широком смысле.


### Упражнение 1.4

Пусть $X,Y$ - независимые гауссовские процессы. Нужно доказать, что $X+Y$ - гауссовский процесс, и найти его матожидание и ковариацию.

**Решение.**

Возьмем произвольный конечный набор моментов $t_1,\ldots,t_n$. Векторы
$$
(X_{t_1},\ldots,X_{t_n}),\qquad (Y_{t_1},\ldots,Y_{t_n})
$$
гауссовские и независимые. Их сумма тоже гауссовский вектор. Значит, все конечномерные распределения процесса $X+Y$ гауссовские, то есть $X+Y$ - гауссовский процесс.

Матожидание:
$$
m_{X+Y}(t)=m_X(t)+m_Y(t).
$$
Ковариация:
$$
K_{X+Y}(t,s)=K_X(t,s)+K_Y(t,s),
$$
потому что из независимости исчезают смешанные ковариации.


### Упражнение 1.5

Нужно доказать, что для гауссовского процесса стационарность в широком смысле влечет стационарность в узком смысле.

**Решение.**

Для гауссовского процесса любое конечномерное распределение полностью определяется вектором матожиданий и ковариационной матрицей.

Если процесс стационарен в широком смысле, то
$$
m(t)=m
$$
не зависит от $t$, а
$$
K(t,s)=R(t-s)
$$
зависит только от разности времен.

Поэтому для любого набора $t_1,\ldots,t_n$ и любого сдвига $h$ векторы
$$
(X_{t_1},\ldots,X_{t_n})
$$
и
$$
(X_{t_1+h},\ldots,X_{t_n+h})
$$
имеют одинаковые векторы матожиданий и одинаковые ковариационные матрицы. Так как оба вектора гауссовские, их распределения совпадают. Это и есть стационарность в узком смысле.



### Упражнение 1.6

Пусть $X$ - гауссовский процесс с нулевым матожиданием и
$$
K(t,s)=ts.
$$
Нужно показать, что почти наверное траектории являются прямыми, выходящими из нуля.

**Решение.**

Для каждого фиксированного $t$ рассмотрим случайную величину
$$
X_t-tX_1.
$$
Ее дисперсия равна
$$
\operatorname{Var}(X_t-tX_1)
=K(t,t)-2tK(t,1)+t^2K(1,1)
=t^2-2t^2+t^2=0.
$$
Значит,
$$
X_t=tX_1
$$
почти наверное для каждого фиксированного $t$.

Строгий технический момент: для несчетного множества индексов нельзя просто пересечь события вероятности $1$ по всем $t\in\mathbb R$. Поэтому корректная формулировка такая: процесс имеет модификацию
$$
\widetilde X_t=tZ,\qquad Z=X_1,
$$
и у этой модификации все траектории являются прямыми, исходящими из нуля.

Почему модификация непрерывна? Для любых $t,s$
$$
\mathbb E|X_t-X_s|^2=t^2+s^2-2ts=(t-s)^2,
$$
поэтому по критерию Колмогорова-Ченцова можно взять непрерывную модификацию. На рациональных $t$ равенство $X_t=tX_1$ выполняется одновременно почти наверное; по непрерывности оно продолжается на все вещественные $t$.



### Упражнение 1.7

Пусть $X$ - гауссовский процесс с нулевым матожиданием и
$$
K(t,s)=f(t)f(s).
$$
Нужно описать вид траекторий.

**Решение.**

Если $f\equiv 0$, то
$$
\operatorname{Var}X_t=0
$$
для всех $t$, значит процесс является нулевым по всем конечномерным распределениям. В стандартной сепарабельной версии можно считать, что почти все траектории тождественно равны нулю.

Пусть теперь существует $t_0$, для которого $f(t_0)\ne0$. Рассмотрим
$$
X_t-\frac{f(t)}{f(t_0)}X_{t_0}.
$$
Его дисперсия:
$$
\operatorname{Var}\left(X_t-\frac{f(t)}{f(t_0)}X_{t_0}\right)
=f(t)^2-2\frac{f(t)}{f(t_0)}f(t)f(t_0)
+\frac{f(t)^2}{f(t_0)^2}f(t_0)^2=0.
$$
Поэтому для каждого фиксированного $t$
$$
X_t=Zf(t),\qquad Z=\frac{X_{t_0}}{f(t_0)}
$$
почти наверное.

Иными словами, все конечномерные распределения совпадают с процессом вида
$$
\widetilde X_t=Zf(t),
$$
где $Z\sim N(0,1)$, если $K(t_0,t_0)=f(t_0)^2$ нормирован именно так. Если требуется утверждение одновременно для всех $t$ на одном множестве $\Omega'$ вероятности $1$, то нужна обычная для гауссовских процессов сепарабельная/регулярная версия; без такой оговорки утверждение для произвольной несчетно индексированной модификации может быть ложно из-за точечных изменений на нулевых множествах.



## Листок 2. Винеровский процесс

### Упражнение 2.1

Пусть $X$ - гауссовский процесс с нулевым матожиданием и
$$
K(t,s)=e^{-|t-s|^2/2}.
$$
Нужно проверить, существует ли модификация с непрерывными траекториями.

**Решение.**

Посчитаем второй момент приращения:
$$
\mathbb E(X_t-X_s)^2
=K(t,t)+K(s,s)-2K(t,s)
=2\left(1-e^{-|t-s|^2/2}\right).
$$
Так как $1-e^{-u}\le u$ при $u\ge0$,
$$
\mathbb E(X_t-X_s)^2\le |t-s|^2.
$$
Приращение $X_t-X_s$ гауссовское, значит для любого $p>1$
$$
\mathbb E|X_t-X_s|^p\le C_p |t-s|^p.
$$
На каждом компакте это удовлетворяет условию Колмогорова-Ченцова, потому что $p=1+(p-1)$ и $p-1>0$. Следовательно, существует модификация с непрерывными траекториями на каждом компакте, а значит и на всей прямой локально.



### Упражнение 2.2

Пусть $X$ - гауссовский процесс с нулевым матожиданием и
$$
K(t,s)=e^{-|t-s|/2}+5(t\wedge s)+21\mathbf 1(t=s).
$$
Нужно проверить, существует ли непрерывная модификация.

**Решение.**

Если бы существовала непрерывная модификация, то при $s\to t$ имели бы $X_s\to X_t$ по вероятности. Для центрированных гауссовских величин это эквивалентно сходимости дисперсии приращения к нулю:
$$
\mathbb E(X_s-X_t)^2\to0.
$$

Но при $s\to t$, $s\ne t$:
$$
\mathbb E(X_t-X_s)^2
=K(t,t)+K(s,s)-2K(t,s)
\to (22+5t)+(22+5t)-2(1+5t)=42.
$$
Среднеквадратичная, а значит и стохастическая непрерывность нарушается. Непрерывной модификации не существует.


### Упражнение 2.3

Пусть $W_t$ - $d$-мерный Винеровский процесс, а $U$ - унитарная матрица. Нужно доказать, что $X_t=UW_t$ тоже $d$-мерный Винеровский процесс.

**Решение.**

Очевидно, $X_0=UW_0=0$, а непрерывность траекторий сохраняется при линейном преобразовании.

Для приращений:
$$
X_t-X_s=U(W_t-W_s).
$$
Так как $W_t-W_s\sim N(0,(t-s)I)$, то
$$
U(W_t-W_s)\sim N(0,(t-s)UU^\ast)=N(0,(t-s)I).
$$
Независимость приращений сохраняется, потому что мы применяем детерминированное линейное преобразование к независимым векторам. Значит, $X$ - $d$-мерный Винеровский процесс.


### Упражнение 2.4

Пусть
$$
B_t=W_t-\frac{t}{T}W_T,\qquad t\in[0,T],
$$
это броуновский мост. Нужно найти
$$
\mathbb E\left[\int_0^T\int_0^T B_tB_s\,dt\,ds\right].
$$

**Решение.**

Ковариация броуновского моста:
$$
\mathbb E[B_tB_s]=t\wedge s-\frac{ts}{T}.
$$
Тогда
$$
\mathbb E\left[\int_0^T\int_0^T B_tB_s\,dt\,ds\right]
=\int_0^T\int_0^T \left(t\wedge s-\frac{ts}{T}\right)\,dt\,ds.
$$
Известно, что
$$
\int_0^T\int_0^T (t\wedge s)\,dt\,ds=\frac{T^3}{3},
$$
а
$$
\int_0^T\int_0^T \frac{ts}{T}\,dt\,ds=\frac1T\left(\frac{T^2}{2}\right)^2=\frac{T^3}{4}.
$$
Итого
$$
\boxed{\frac{T^3}{12}}.
$$


### Упражнение 2.5

Пусть $W$ - одномерный Винеровский процесс, $c>0$. Нужно доказать, что
$$
X_t=\sqrt c\,W_{t/c}
$$
тоже Винеровский процесс.

**Решение.**

$X_0=0$, траектории непрерывны. Для $0\le s<t$:
$$
X_t-X_s=\sqrt c\,(W_{t/c}-W_{s/c}).
$$
Это гауссовская случайная величина с нулевым средним и дисперсией
$$
c\left(\frac{t}{c}-\frac{s}{c}\right)=t-s.
$$
Независимость приращений сохраняется, потому что интервалы времени после преобразования $u\mapsto u/c$ остаются непересекающимися. Значит, $X$ - Винеровский процесс.


### Упражнение 2.6

Пусть $W$ - Винеровский процесс с непрерывными траекториями. Используя
$$
\frac{W_t}{t}\to 0 \quad \text{п.н. при } t\to\infty,
$$
нужно доказать, что $Y_t=tW_{1/t}$ - тоже Винеровский процесс с непрерывными траекториями.

**Решение.**

Для $t>0$ процесс $Y_t=tW_{1/t}$ гауссовский и имеет нулевое матожидание. Его ковариация:
$$
\mathbb E[Y_tY_s]=ts\,\mathbb E[W_{1/t}W_{1/s}]
=ts\left(\frac1t\wedge \frac1s\right)=t\wedge s.
$$
То есть конечномерные распределения совпадают с конечномерными распределениями Винеровского процесса.

На $(0,\infty)$ непрерывность следует из непрерывности $W$. В нуле определим $Y_0=0$. При $t\downarrow0$ положим $u=1/t\to\infty$:
$$
Y_t=tW_{1/t}=\frac{W_u}{u}\to0
$$
почти наверное по усиленному закону больших чисел для Винеровского процесса. Значит, есть непрерывная версия, и $Y$ - Винеровский процесс.


### Упражнение 2.7

Пусть
$$
Y_t=\frac1t\int_0^t W_s\,ds.
$$
Нужно доказать, что это гауссовский процесс, найти матожидание и ковариационную функцию, проверить стационарность в широком смысле.

**Решение.**

Интеграл можно понимать как предел интегральных сумм. Каждая интегральная сумма является линейной комбинацией гауссовских случайных величин, значит гауссовская. Предел в $L^2$ также гауссовский, поэтому $Y$ - гауссовский процесс.

Матожидание:
$$
\mathbb E Y_t=\frac1t\int_0^t\mathbb E W_s\,ds=0.
$$

Ковариация:
$$
\operatorname{Cov}(Y_t,Y_s)
=\frac1{ts}\int_0^t\int_0^s (u\wedge v)\,du\,dv.
$$
Пусть для определенности $t\le s$. Тогда
$$
\int_0^t\int_0^s (u\wedge v)\,dv\,du
=\int_0^t\left(\frac{u^2}{2}+u(s-u)\right)du
=\frac{st^2}{2}-\frac{t^3}{6}.
$$
Значит,
$$
\operatorname{Cov}(Y_t,Y_s)=\frac t2-\frac{t^2}{6s},\qquad t\le s.
$$
В симметричной форме:
$$
K(t,s)=\frac{\min(t,s)}2-\frac{\min(t,s)^2}{6\max(t,s)}.
$$
Дисперсия:
$$
K(t,t)=\frac t3,
$$
она зависит от $t$. Поэтому процесс не стационарен в широком смысле.


### Упражнение 2.8, бонус

Нужно доказать, что для Винеровского процесса
$$
\frac{W_t}{t}\to0
$$
почти наверное при $t\to\infty$.

**Решение.**

Сначала докажем сходимость вдоль целых моментов. Так как
$$
W_n=\sum_{k=1}^n (W_k-W_{k-1}),
$$
а приращения $W_k-W_{k-1}$ независимы, одинаково распределены, имеют матожидание $0$, то по усиленному закону больших чисел:
$$
\frac{W_n}{n}\to0
$$
почти наверное.

Осталось контролировать промежутки $[n,n+1]$. Для любого $\varepsilon>0$:
$$
\mathbb P\left(\sup_{t\in[n,n+1]}|W_t-W_n|>\varepsilon n\right)
=\mathbb P\left(\sup_{u\in[0,1]}|W_{n+u}-W_n|>\varepsilon n\right).
$$
По стационарности приращений это равно
$$
\mathbb P\left(\sup_{u\in[0,1]}|W_u|>\varepsilon n\right),
$$
а эта вероятность убывает экспоненциально по $n^2$; например, это следует из принципа отражения и гауссовской оценки хвоста. Поэтому ряд таких вероятностей сходится. По лемме Бореля-Кантелли почти наверное начиная с некоторого номера
$$
\sup_{t\in[n,n+1]}|W_t-W_n|\le \varepsilon n.
$$

Если $t\in[n,n+1]$, то
$$
\left|\frac{W_t}{t}\right|
\le \left|\frac{W_n}{n}\right|+\frac{\sup_{u\in[n,n+1]}|W_u-W_n|}{n}.
$$
Правая часть почти наверное стремится к величине не больше $\varepsilon$. Так как $\varepsilon>0$ произвольно, получаем
$$
\frac{W_t}{t}\to0
$$
почти наверное.


## Вычислительная часть. Оценка параметров в модели Блэка-Шоулза

Геометрическое броуновское движение задается формулой
$$
X_t=X_0\exp\left((\mu-\sigma^2/2)t+\sigma W_t\right).
$$

Пусть наблюдения даны в моменты $t_0,\ldots,t_n$, а
$$
Y_k=\ln X_{t_k}-\ln X_{t_{k-1}},\qquad h_k=t_k-t_{k-1}.
$$
Тогда
$$
Y_k\sim N\left((\mu-\sigma^2/2)h_k,\sigma^2 h_k\right),
$$
и эти приращения независимы.



### Функция правдоподобия и оценки

Функция правдоподобия:
$$
L(\mu,\sigma)=\prod_{k=1}^n
\frac1{\sqrt{2\pi\sigma^2h_k}}
\exp\left(
-\frac{(Y_k-(\mu-\sigma^2/2)h_k)^2}{2\sigma^2h_k}
\right).
$$

Логарифм:
$$
\ell(\mu,\sigma)
=-n\log\sigma-\frac12\sum_{k=1}^n\log(2\pi h_k)
-\sum_{k=1}^n\frac{(Y_k-(\mu-\sigma^2/2)h_k)^2}{2\sigma^2h_k}.
$$

Обозначим
$$
r_k(\mu,\sigma)=Y_k-(\mu-\sigma^2/2)h_k.
$$
Тогда настоящие производные по исходным параметрам $\mu,\sigma$:
$$
\partial_\mu\ell
=\frac1{\sigma^2}\sum_{k=1}^n r_k(\mu,\sigma),
$$
$$
\partial_\sigma\ell
=-\frac n\sigma-\frac1\sigma\sum_{k=1}^n r_k(\mu,\sigma)
+\frac1{\sigma^3}\sum_{k=1}^n\frac{r_k(\mu,\sigma)^2}{h_k}.
$$

Для нахождения максимума удобнее сделать замену
$$
\theta=\mu-\frac{\sigma^2}{2}.
$$
Это взаимно однозначная параметризация при $\sigma>0$, поэтому максимум в $(\theta,\sigma)$ дает максимум и в $(\mu,\sigma)$. В этих параметрах
$$
\ell(\theta,\sigma)
=-n\log\sigma-\frac12\sum_{k=1}^n\log(2\pi h_k)
-\frac1{2\sigma^2}\sum_{k=1}^n\frac{(Y_k-\theta h_k)^2}{h_k},
$$
и
$$
\partial_\theta\ell
=\frac1{\sigma^2}\sum_{k=1}^n(Y_k-\theta h_k),
$$
$$
\partial_\sigma\ell
=-\frac n\sigma+\frac1{\sigma^3}\sum_{k=1}^n\frac{(Y_k-\theta h_k)^2}{h_k}.
$$

Приравнивая производные к нулю, получаем
$$
\widehat\theta=\frac{\sum_{k=1}^nY_k}{\sum_{k=1}^nh_k},
$$
$$
\widehat\sigma^2=\frac1n\sum_{k=1}^n\frac{(Y_k-\widehat\theta h_k)^2}{h_k},
$$
и возвращаемся к исходному параметру:
$$
\boxed{\widehat\mu=\widehat\theta+\frac{\widehat\sigma^2}{2}},\qquad
\boxed{\widehat\sigma=\sqrt{\widehat\sigma^2}}.
$$

Если сетка равномерная, $h_k\equiv h$, то это особенно прозрачно:
$$
\widehat\theta=\frac1{nh}\sum_{k=1}^nY_k,\qquad
\widehat\sigma^2=\frac1{nh}\sum_{k=1}^n(Y_k-\widehat\theta h)^2.
$$


```python
import pickle
from itertools import combinations_with_replacement
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
plt.style.use("seaborn-v0_8-whitegrid")
```

```python

def simulateWienerProcess(x0, ts, Ntraj):
    # Simulates trajectories of a one-dimensional Wiener process.
    # Returns an array of shape (Ntraj, len(ts)).
    ts = np.asarray(ts, dtype=float)
    if np.any(np.diff(ts) < 0):
        raise ValueError("Time grid must be nondecreasing")
    xs = np.zeros((Ntraj, len(ts)))
    xs[:, 0] = x0
    dts = np.diff(ts)
    increments = np.random.normal(0.0, np.sqrt(dts), size=(Ntraj, len(ts) - 1))
    xs[:, 1:] = x0 + np.cumsum(increments, axis=1)
    return xs


class GBMModel:
    # Geometric Brownian Motion with parameter estimation and simulation.

    def __init__(self, mu=0.0, sigma=1.0):
        self.mu = float(mu)
        self.sigma = float(sigma)

    def fit(self, xs, ts):
        xs = np.asarray(xs, dtype=float)
        ts = np.asarray(ts, dtype=float)
        if np.any(xs <= 0):
            raise ValueError("GBM observations must be strictly positive")
        if xs.shape != ts.shape:
            raise ValueError("xs and ts must have the same shape")
        hs = np.diff(ts)
        if np.any(hs <= 0):
            raise ValueError("Observation times must be strictly increasing")

        log_returns = np.diff(np.log(xs))
        theta_hat = np.sum(log_returns) / np.sum(hs)
        sigma2_hat = np.mean((log_returns - theta_hat * hs) ** 2 / hs)

        self.sigma = float(np.sqrt(max(sigma2_hat, 0.0)))
        self.mu = float(theta_hat + 0.5 * sigma2_hat)
        return self

    def simulate(self, x0, ts, Ntraj):
        if x0 <= 0:
            raise ValueError("Initial GBM value must be strictly positive")
        ts = np.asarray(ts, dtype=float)
        rel_ts = ts - ts[0]
        w = simulateWienerProcess(0.0, rel_ts, Ntraj)
        exponent = (self.mu - 0.5 * self.sigma**2) * rel_ts[None, :] + self.sigma * w
        return float(x0) * np.exp(exponent)

    def mean(self, x0, ts):
        ts = np.asarray(ts, dtype=float)
        rel_ts = ts - ts[0]
        return float(x0) * np.exp(self.mu * rel_ts)

    def confidence_interval(self, x0, ts, z=1.96):
        ts = np.asarray(ts, dtype=float)
        rel_ts = ts - ts[0]
        center = (self.mu - 0.5 * self.sigma**2) * rel_ts
        width = z * self.sigma * np.sqrt(rel_ts)
        low = float(x0) * np.exp(center - width)
        high = float(x0) * np.exp(center + width)
        return low, high
```

## Часть 5. Оценка американского опциона

Рассматриваем бермудскую дискретизацию американского опциона. В моменты $t_k$ сравниваем немедленную выплату $f(X_{t_k})$ и продолжение:
$$
C_k(x)=e^{-r(t_{k+1}-t_k)}\mathbb E[V_{t_{k+1}}(X_{t_{k+1}})\mid X_{t_k}=x].
$$
Если $f(x)\ge C_k(x)$, исполняем опцион, иначе продолжаем. Условное матожидание приближаем полиномиальной регрессией по методу Longstaff-Schwartz.

Практически важная деталь: регрессия может дать отрицательное значение продолжения в областях, где истинная цена продолжения неотрицательна. Поэтому в реализации ниже мы разрешаем раннее исполнение только для траекторий, где payoff строго положителен. Это стандартная защита от искусственного "исполнения за ноль" из-за регрессионной ошибки.


```python
def simulateWienerProcessMultivar(x0, ts, Ntraj):
    # Simulates trajectories of a d-dimensional Wiener process.
    # Returns an array of shape (Ntraj, d, len(ts)).
    x0 = np.asarray(x0, dtype=float)
    ts = np.asarray(ts, dtype=float)
    d = x0.shape[0]
    xs = np.zeros((Ntraj, d, len(ts)))
    xs[:, :, 0] = x0[None, :]
    dts = np.diff(ts)
    increments = np.random.normal(0.0, np.sqrt(dts)[None, None, :], size=(Ntraj, d, len(ts) - 1))
    xs[:, :, 1:] = x0[None, :, None] + np.cumsum(increments, axis=2)
    return xs


class GBMModelMultivar:
    # Multidimensional GBM with independent Brownian components.

    def __init__(self, mu=0, sigma=1):
        self.mu = np.asarray(mu, dtype=float)
        self.sigma = np.asarray(sigma, dtype=float)

    def simulate(self, x0, ts, Ntraj):
        x0 = np.asarray(x0, dtype=float)
        ts = np.asarray(ts, dtype=float)
        rel_ts = ts - ts[0]
        w = simulateWienerProcessMultivar(np.zeros_like(x0), rel_ts, Ntraj)
        drift = (self.mu - 0.5 * self.sigma**2)[None, :, None] * rel_ts[None, None, :]
        diffusion = self.sigma[None, :, None] * w
        return x0[None, :, None] * np.exp(drift + diffusion)


class PolynomialFeatureMap:
    def __init__(self, degree=2):
        self.degree = degree
        self.combinations = None

    def fit(self, X):
        X = np.asarray(X, dtype=float)
        d = X.shape[1]
        self.combinations = [()]
        for deg in range(1, self.degree + 1):
            self.combinations.extend(combinations_with_replacement(range(d), deg))
        return self

    def transform(self, X):
        X = np.asarray(X, dtype=float)
        if self.combinations is None:
            self.fit(X)
        cols = []
        for combo in self.combinations:
            if len(combo) == 0:
                cols.append(np.ones(X.shape[0]))
            else:
                cols.append(np.prod(X[:, combo], axis=1))
        return np.column_stack(cols)

    def fit_transform(self, X):
        return self.fit(X).transform(X)


class ValueFunction:
    def __init__(self, deg=2):
        self.coef_ = None
        self.featureTransform = PolynomialFeatureMap(degree=deg)

    def fit(self, X, y):
        X_poly = self.featureTransform.fit_transform(np.asarray(X, dtype=float))
        self.coef_, *_ = np.linalg.lstsq(X_poly, np.asarray(y, dtype=float).ravel(), rcond=None)
        return self

    def predict(self, X):
        X_poly = self.featureTransform.transform(np.asarray(X, dtype=float))
        return X_poly @ self.coef_

    def getParameters(self):
        return self.coef_
```

```python
vf = ValueFunction(deg=3)

x_grid = np.arange(0, 1, 1 / 20)[:, None]
y_grid = 0.2 * x_grid[:, 0] ** 3 + 0.8 * x_grid[:, 0] ** 2 - 2 * x_grid[:, 0] + 3

vf.fit(x_grid, y_grid)
print("Polynomial regression coefficients:", np.round(vf.getParameters(), 4))
```

```python

class OptionPricer:
    # Bermudan/American option pricer via Longstaff-Schwartz backward induction.

    def __init__(self, priceModel, payment, executionTimes, discountFunction, x0):
        self.priceModel = priceModel
        self.payment = payment
        self.executionTimes = np.asarray(executionTimes, dtype=float)
        self.discountFunction = discountFunction
        self.x0 = np.asarray(x0, dtype=float)
        self.continuationModels = [ValueFunction(deg=2) for _ in range(len(executionTimes) - 2)]

    def _fit_continuation(self, model, X, y, immediate):
        # Longstaff-Schwartz is usually fitted on in-the-money paths.
        in_the_money = immediate > 0
        if np.sum(in_the_money) >= 10:
            model.fit(X[in_the_money], y[in_the_money])
        else:
            model.fit(X, y)

    def computeHighBiasEstimator(self, Ntrajs):
        paths = self.priceModel.simulate(self.x0, self.executionTimes, Ntrajs)
        p = len(self.executionTimes) - 1
        values = self.payment(paths[:, :, p])

        for k in range(p - 1, 0, -1):
            dt = self.executionTimes[k + 1] - self.executionTimes[k]
            discounted_next = self.discountFunction(dt) * values
            Xk = paths[:, :, k]
            immediate = self.payment(Xk)
            model = self.continuationModels[k - 1]
            self._fit_continuation(model, Xk, discounted_next, immediate)
            continuation = model.predict(Xk)
            exercise = (immediate > 0) & (immediate >= continuation)
            values = np.where(exercise, immediate, discounted_next)

        first_dt = self.executionTimes[1] - self.executionTimes[0]
        samples = self.discountFunction(first_dt) * values
        immediate0 = float(self.payment(self.x0[None, :])[0])
        if immediate0 > float(np.mean(samples)):
            samples = np.full(Ntrajs, immediate0)

        priceMean = float(np.mean(samples))
        priceStd = float(np.std(samples, ddof=1) / np.sqrt(Ntrajs))
        return priceMean, priceStd

    def computeLowEstimator(self, Ntrajs):
        paths = self.priceModel.simulate(self.x0, self.executionTimes, Ntrajs)
        p = len(self.executionTimes) - 1
        rewards = np.zeros(Ntrajs)
        alive = np.ones(Ntrajs, dtype=bool)

        for k in range(1, p):
            Xk = paths[:, :, k]
            immediate = self.payment(Xk)
            continuation = self.continuationModels[k - 1].predict(Xk)
            exercise = alive & (immediate > 0) & (immediate >= continuation)
            rewards[exercise] = self.discountFunction(self.executionTimes[k]) * immediate[exercise]
            alive[exercise] = False

        if np.any(alive):
            terminal = self.payment(paths[alive, :, p])
            rewards[alive] = self.discountFunction(self.executionTimes[p]) * terminal

        immediate0 = float(self.payment(self.x0[None, :])[0])
        if immediate0 > float(np.mean(rewards)):
            rewards = np.full(Ntrajs, immediate0)

        priceLow = float(np.mean(rewards))
        priceLowStd = float(np.std(rewards, ddof=1) / np.sqrt(Ntrajs))
        return priceLow, priceLowStd
```


### Эксперименты

Параметры: $r=0.05$, $\delta=0.1$, $\sigma=0.2$, $T=3$, шаг $0.3$, страйк $K=100$.
Выплата:
$$
f(x)=(\max_i x_i-K)_+.
$$
Это payoff call-опциона на максимум из двух активов. В тексте задания ниже встречается слово `max-put`, но формула выплаты однозначно соответствует max-call, поэтому в вычислениях используем именно ее.

Сравним три начальные точки: $[100,100]$, $[110,110]$, $[90,90]$.


```python
a = 0
b = 3
Nt = 10
h = (b - a) / Nt
ts_option = np.arange(a, b + h / 2, h)

r = 0.05
delta = 0.1
mu_vec = (r - delta) * np.ones(2)
sigma_vec = 0.2 * np.ones(2)
K = 100

gbm_option = GBMModelMultivar(mu_vec, sigma_vec)


def basketPayment(x):
    x = np.asarray(x, dtype=float)
    return np.maximum(np.max(x, axis=1) - K, 0.0)


def expDiscount(h):
    return np.exp(-r * h)


def european_price_mc(x0, N=100_000):
    paths = gbm_option.simulate(np.asarray(x0, dtype=float), ts_option, N)
    payoff = basketPayment(paths[:, :, -1])
    discounted = expDiscount(ts_option[-1]) * payoff
    return float(np.mean(discounted)), float(np.std(discounted, ddof=1) / np.sqrt(N))
```

```python

np.random.seed(2026)

N_train_grid = [500, 1000, 2000, 4000, 8000, 16000]
N_test = 20_000
scenarios = {
    "x0 = [100, 100]": np.array([100, 100]),
    "x0 = [110, 110]": np.array([110, 110]),
    "x0 = [90, 90]": np.array([90, 90]),
}

experiment_results = {}

for label, x0_scenario in scenarios.items():
    rows = []
    print(label)
    for N_train in N_train_grid:
        pricer = OptionPricer(gbm_option, basketPayment, ts_option, expDiscount, x0_scenario)
        high_price, high_se = pricer.computeHighBiasEstimator(N_train)
        low_price, low_se = pricer.computeLowEstimator(N_test)
        rows.append((N_train, high_price, high_se, low_price, low_se))
        print(
            f"  N={N_train:5d}: high={high_price:8.4f} +/- {1.96 * high_se:.4f}, "
            f"low={low_price:8.4f} +/- {1.96 * low_se:.4f}"
        )
    euro_price, euro_se = european_price_mc(x0_scenario)
    experiment_results[label] = {
        "x0": x0_scenario,
        "rows": np.array(rows, dtype=float),
        "european": (euro_price, euro_se),
    }
    print(f"  European MC: {euro_price:.4f} +/- {1.96 * euro_se:.4f}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

for ax, (label, result) in zip(axes, experiment_results.items()):
    rows = result["rows"]
    Ns = rows[:, 0]
    high = rows[:, 1]
    high_ci = 1.96 * rows[:, 2]
    low = rows[:, 3]
    low_ci = 1.96 * rows[:, 4]
    euro, euro_se = result["european"]

    ax.set_title(label)
    ax.set_xscale("log", base=2)
    ax.errorbar(Ns, high, yerr=high_ci, marker="o", label="high-bias estimator")
    ax.errorbar(Ns, low, yerr=low_ci, marker="s", label="low estimator")
    ax.axhline(euro, color="tab:green", linestyle="--", label="European MC")
    ax.fill_between(Ns, euro - 1.96 * euro_se, euro + 1.96 * euro_se, color="tab:green", alpha=0.12)
    ax.set_xlabel("number of training paths")
    ax.grid(True, which="both", alpha=0.35)

axes[0].set_ylabel("option price")
axes[0].legend(loc="upper left")
plt.tight_layout()
plt.show()
```


**Вывод по экспериментам.**

Самый дорогой опцион получается при $X_0=[110,110]$, затем при $X_0=[100,100]$, самый дешевый - при $X_0=[90,90]$. Это естественно: выплата равна $(\max_i X_i-K)_+$, поэтому чем выше стартовые цены относительно страйка, тем больше вероятность оказаться в деньгах и тем выше текущая стоимость права исполнения.

Нижняя оценка обычно ниже оценки на обучающей выборке, потому что на тестовых траекториях мы используем уже зафиксированное правило остановки и не подгоняемся под те же данные. На конечном Монте-Карло возможны небольшие пересечения доверительных интервалов; это не противоречит идее метода, а показывает шум регрессии и симуляции.

Европейский опцион дешевле американского/бермудского, потому что европейский можно исполнить только в момент $T$, а бермудский дает дополнительную свободу выбрать один из моментов исполнения. При наличии дивидендов раннее исполнение call-опциона может быть выгодным, поэтому разница здесь содержательно ожидаема.



## Регион исполнения

Нарисуем на плоскости $(x_1,x_2)$ знак разности
$$
C_k(x)-f(x),
$$
где $C_k(x)$ - оцененная стоимость продолжения. Если $C_k(x)-f(x)>0$, выгоднее ждать. Если $C_k(x)-f(x)<0$, выгоднее исполнять. Граница исполнения приближенно задается уровнем $0$.

Важно интерпретировать эти картинки качественно: регрессия обучалась на облаке симулированных точек, поэтому далеко от типичных значений траекторий возможна экстраполяция.


```python
x0_boundary = np.array([100, 100])
boundary_pricer = OptionPricer(gbm_option, basketPayment, ts_option, expDiscount, x0_boundary)
boundary_pricer.computeHighBiasEstimator(16000)

grid = np.linspace(60, 160, 90)
xx, yy = np.meshgrid(grid, grid)
grid_points = np.column_stack([xx.ravel(), yy.ravel()])
payoff_grid = basketPayment(grid_points).reshape(xx.shape)

selected_k = [1, 3, 6, 9]

fig, axes = plt.subplots(2, 2, figsize=(12, 10), sharex=True, sharey=True)

for ax, k in zip(axes.ravel(), selected_k):
    continuation = boundary_pricer.continuationModels[k - 1].predict(grid_points).reshape(xx.shape)
    diff = continuation - payoff_grid
    cf = ax.contourf(xx, yy, diff, levels=30, cmap="coolwarm", alpha=0.85)
    ax.contour(xx, yy, diff, levels=[0], colors="black", linewidths=2)
    ax.contour(xx, yy, payoff_grid, levels=[0], colors="gray", linestyles="--", linewidths=1)
    ax.set_title(f"t = {ts_option[k]:.1f}: continuation - payoff")
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")

fig.colorbar(cf, ax=axes.ravel().tolist(), shrink=0.85)
plt.show()
```

**Интерпретация области исполнения.**

Область ожидания соответствует положительным значениям $C_k(x)-f(x)$: модель считает, что продолжение в среднем ценнее немедленной выплаты. Область исполнения соответствует отрицательным значениям: текущая выплата уже достаточно хороша, и ждать менее выгодно.

Для max-call опциона выплата растет, когда хотя бы один актив сильно выше страйка. Но из-за дивидендов активы имеют отрицательный риск-нейтральный снос $\mu=r-\delta<0$, поэтому при больших текущих ценах ожидание может стать менее привлекательным: есть риск потерять уже доступную внутреннюю стоимость. Ближе к сроку истечения область исполнения расширяется, потому что остается меньше времени, чтобы будущая случайность компенсировала отказ от текущей выплаты.

Небольшие неровности и странные изгибы границы - нормальный эффект приближенной регрессии и Монте-Карло шума. Сама идея правила при этом читается правильно: ждать выгодно там, где текущая выплата мала или есть существенная опционная ценность будущего роста; исполнять выгодно там, где текущая внутренняя стоимость уже велика, а отрицательный снос из-за дивидендов делает ожидание дорогим.
