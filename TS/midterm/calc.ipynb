преобразуем датасет в удобный формат

```python
FILEPATH = "/Users/maria/Desktop/Code/HSE/TS/midterm/API_SL.SRV.EMPL.ZS_DS2_en_csv_v2_4736.csv"

COUNTRIES = ["Germany", "Turkiye"]
YEAR_START = 1991
YEAR_END = 2025

# 1. Чтение WB csv
raw_df = pd.read_csv(FILEPATH, skiprows=4)

# 2. Убираем мусорный хвост, если есть
raw_df = raw_df.loc[:, ~raw_df.columns.str.contains(r"^Unnamed")]

# 3. Оставляем только нужные служебные колонки и годы
year_cols = [str(y) for y in range(YEAR_START, YEAR_END + 1)]
base_cols = ["Country Name", "Country Code", "Indicator Name", "Indicator Code"]
keep_cols = [c for c in base_cols + year_cols if c in raw_df.columns]

df = raw_df[keep_cols].copy()

# 4. Берем только две страны
df = df[df["Country Name"].isin(COUNTRIES)].copy()

# 5. Проверка, что обе страны нашлись
found_countries = set(df["Country Name"].unique())
missing_countries = set(COUNTRIES) - found_countries
if missing_countries:
    raise ValueError(f"Не найдены страны: {missing_countries}")

# 6. Приводим годовые столбцы к numeric
for col in year_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# 7. WIDE формат: index = year, columns = countries
df_wide = (
    df.set_index("Country Name")[year_cols]
      .T
      .rename_axis("year")
)

df_wide.index = df_wide.index.astype(int)
df_wide = df_wide.sort_index()

# 8. LONG формат: year, country, value
df_long = (
    df_wide.reset_index()
           .melt(id_vars="year", var_name="country", value_name="value")
           .sort_values(["country", "year"])
           .reset_index(drop=True)
)

# 9. Метаданные в отдельной табличке
df_meta = (
    df[["Country Name", "Country Code", "Indicator Name", "Indicator Code"]]
    .drop_duplicates()
    .sort_values("Country Name")
    .reset_index(drop=True)
)

# 10. Быстрые проверки
print("META")
display(df_meta)

print("WIDE")
display(df_wide.head())

print("LONG")
display(df_long.head())

print("Число наблюдений по странам:")
display(df_long.groupby("country")["value"].apply(lambda s: s.notna().sum()))
```

## Итерация 2. Стационарность и порядок интегрируемости

Во второй итерации проверим стационарность рядов Germany и Turkiye, рассмотрим коррелограммы и определим необходимый порядок дифференцирования.

```python
import warnings
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.stattools import adfuller, kpss
from statsmodels.tools.sm_exceptions import InterpolationWarning

warnings.filterwarnings("ignore", category=InterpolationWarning)

try:
    from statsmodels.tsa.stattools import phillips_perron
    phillips_perron_available = True
except ImportError:
    phillips_perron = None
    phillips_perron_available = False

series_dict = {
    "Germany": df_wide["Germany"].dropna().astype(float),
    "Turkiye": df_wide["Turkiye"].dropna().astype(float)
}

country_colors = {
    "Germany": globals().get("germany_color", "#4C78A8"),
    "Turkiye": globals().get("turkiye_color", "#C46A3A")
}

grid_color = globals().get("grid_color", "#D9E0E8")
spine_color = globals().get("spine_color", "#C7D0DA")
text_color = globals().get("text_color", "#27313F")

if "style_report_axes" not in globals():
    def style_report_axes(ax):
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        ax.spines["left"].set_color(spine_color)
        ax.spines["bottom"].set_color(spine_color)
        ax.spines["left"].set_linewidth(0.8)
        ax.spines["bottom"].set_linewidth(0.8)
        ax.grid(axis="y", color=grid_color, linewidth=0.8, alpha=0.85)
        ax.grid(axis="x", color=grid_color, linewidth=0.6, alpha=0.4)
        ax.set_axisbelow(True)
        ax.tick_params(axis="both", labelsize=11, length=0, pad=6)
        ax.margins(x=0.015)

def format_p_value(p_value):
    if p_value < 0.001:
        return "< 0.001"
    return f"{p_value:.3f}"

def interpret_test(test_name, p_value, kpss_regression=None):
    if test_name.startswith("ADF") or test_name.startswith("Phillips-Perron"):
        return "H0 отвергается" if p_value < 0.05 else "H0 не отвергается"
    if kpss_regression == "ct":
        return "H0 не отвергается" if p_value >= 0.05 else "H0 отвергается"
    return "H0 не отвергается" if p_value >= 0.05 else "H0 отвергается"

def build_test_table(series_map, stage_label, adf_regression, kpss_regression):
    rows = []
    for country, series in series_map.items():
        adf_stat, adf_pvalue = adfuller(series, regression=adf_regression, autolag="AIC")[:2]
        rows.append({
            "stage": stage_label,
            "country": country,
            "test": f"ADF ({adf_regression})",
            "statistic": round(float(adf_stat), 4),
            "p-value": float(adf_pvalue),
            "вывод": interpret_test(f"ADF ({adf_regression})", float(adf_pvalue))
        })

        kpss_stat, kpss_pvalue = kpss(series, regression=kpss_regression, nlags="auto")[:2]
        rows.append({
            "stage": stage_label,
            "country": country,
            "test": f"KPSS ({kpss_regression})",
            "statistic": round(float(kpss_stat), 4),
            "p-value": float(kpss_pvalue),
            "вывод": interpret_test(f"KPSS ({kpss_regression})", float(kpss_pvalue), kpss_regression=kpss_regression)
        })

        if phillips_perron_available:
            pp_stat, pp_pvalue = phillips_perron(series)[:2]
            rows.append({
                "stage": stage_label,
                "country": country,
                "test": "Phillips-Perron",
                "statistic": round(float(pp_stat), 4),
                "p-value": float(pp_pvalue),
                "вывод": interpret_test("Phillips-Perron", float(pp_pvalue))
            })

    table = pd.DataFrame(rows)
    table["p-value"] = table["p-value"].apply(format_p_value)
    return table
```

Сначала посмотрим на ACF и PACF исходных рядов, чтобы оценить характер зависимости и наличие признаков нестационарности.

```python
lags = 10
fig, axes = plt.subplots(2, 2, figsize=(12.5, 8.2), dpi=160)

for row, country in enumerate(["Germany", "Turkiye"]):
    color = country_colors[country]
    plot_acf(series_dict[country], lags=lags, ax=axes[row, 0], alpha=0.05, color=color, title="")
    plot_pacf(series_dict[country], lags=lags, ax=axes[row, 1], alpha=0.05, method="ywmle", color=color, title="")

    axes[row, 0].set_title(f"ACF: {country}", pad=12, color=text_color, fontweight="semibold")
    axes[row, 1].set_title(f"PACF: {country}", pad=12, color=text_color, fontweight="semibold")
    axes[row, 0].set_xlabel("Лаг")
    axes[row, 1].set_xlabel("Лаг")
    axes[row, 0].set_ylabel("Корреляция")
    axes[row, 1].set_ylabel("Частная корреляция")

    style_report_axes(axes[row, 0])
    style_report_axes(axes[row, 1])

fig.suptitle("ACF и PACF исходных рядов", fontsize=15, color=text_color, fontweight="semibold", y=1.02)
fig.tight_layout()
plt.show()
```

- Для Germany ACF затухает медленно, а PACF дает крупный первый пик и затем заметно слабее, что хорошо согласуется с наличием тренда и нестационарности уровня ряда.
- Для Turkiye картина похожая: ACF убывает медленно, а высокий первый пик PACF указывает на сильную инерцию и тоже не поддерживает гипотезу стационарности уровня.

Теперь проведем формальные тесты стационарности на уровнях ряда. Для исходных рядов используем спецификацию с трендом, поскольку в динамике заметен выраженный восходящий тренд.

```python
level_stationarity_tests = build_test_table(
    series_map=series_dict,
    stage_label="Уровень ряда",
    adf_regression="ct",
    kpss_regression="ct"
)

display(level_stationarity_tests[["country", "test", "statistic", "p-value", "вывод"]])

if not phillips_perron_available:
    print("Phillips-Perron недоступен в установленной версии statsmodels, поэтому тест не включен в таблицу.")
```

- H0 теста ADF: в ряду есть единичный корень, то есть ряд нестационарен.
- H0 теста KPSS: ряд стационарен относительно выбранной детерминированной компоненты; здесь на уровнях проверяется тренд-стационарность.
- H0 теста Phillips-Perron: в ряду есть единичный корень. В текущей версии `statsmodels` этот тест недоступен, поэтому дальше используем ADF и KPSS.
- На уровнях оба ряда выглядят нестационарными: для Germany ADF(ct) дает `p = 0.905`, KPSS(ct) дает `p = 0.012`; для Turkiye ADF(ct) дает `p = 0.755`, KPSS(ct) дает `p = 0.047`. В обоих случаях уровень ряда не следует считать стационарным.

Далее перейдем к первым разностям, чтобы проверить, снимается ли тренд после однократного дифференцирования.

```python
diff1_series_dict = {country: series.diff().dropna() for country, series in series_dict.items()}

fig, axes = plt.subplots(2, 3, figsize=(15, 8.5), dpi=160)

for row, country in enumerate(["Germany", "Turkiye"]):
    color = country_colors[country]
    diff_series = diff1_series_dict[country]

    axes[row, 0].plot(diff_series.index, diff_series.values, color=color, linewidth=2.5, alpha=0.9, solid_capstyle="round")
    axes[row, 0].axhline(0, color="#7A8794", linewidth=1.0, alpha=0.8)
    axes[row, 0].set_title(f"Первая разность: {country}", pad=12, color=text_color, fontweight="semibold")
    axes[row, 0].set_xlabel("Год")
    axes[row, 0].set_ylabel("Изменение показателя")
    style_report_axes(axes[row, 0])

    plot_acf(diff_series, lags=lags, ax=axes[row, 1], alpha=0.05, color=color, title="")
    axes[row, 1].set_title(f"ACF первой разности: {country}", pad=12, color=text_color, fontweight="semibold")
    axes[row, 1].set_xlabel("Лаг")
    axes[row, 1].set_ylabel("Корреляция")
    style_report_axes(axes[row, 1])

    plot_pacf(diff_series, lags=lags, ax=axes[row, 2], alpha=0.05, method="ywmle", color=color, title="")
    axes[row, 2].set_title(f"PACF первой разности: {country}", pad=12, color=text_color, fontweight="semibold")
    axes[row, 2].set_xlabel("Лаг")
    axes[row, 2].set_ylabel("Частная корреляция")
    style_report_axes(axes[row, 2])

fig.suptitle("Первые разности и их коррелограммы", fontsize=15, color=text_color, fontweight="semibold", y=1.02)
fig.tight_layout()
plt.show()
```

```python
diff1_stationarity_tests = build_test_table(
    series_map=diff1_series_dict,
    stage_label="Первая разность",
    adf_regression="c",
    kpss_regression="c"
)

display(diff1_stationarity_tests[["country", "test", "statistic", "p-value", "вывод"]])
```

- После первой разности обе ACF становятся заметно короче, а визуальный тренд исчезает, что уже выглядит как движение к стационарности.
- Для Turkiye первая разность можно считать стационарной достаточно уверенно: ADF(c) дает `p < 0.001`, KPSS(c) дает `p = 0.100`.
- Для Germany сигнал смешанный: ADF(c) дает `p < 0.001`, но KPSS(c) дает `p = 0.018`. Поэтому для Germany полезно дополнительно посмотреть вторую разность и проверить, не возникнет ли признаков передифференцирования.

Для Germany дополнительно рассмотрим вторую разность, поскольку после первой разности формальные тесты дают не полностью однозначный результат.

## 3. Анализ тренда и структурных сдвигов

### 3.1. Основные компоненты временного ряда

Сначала уточним, как выглядит трендовая компонента в двух рядах, а затем сравним их коррелограммы для исходных уровней.

```python
import statsmodels.api as sm
from scipy.stats import kendalltau, linregress, f
from statsmodels.sandbox.stats.runs import runstest_1samp

if "series_dict" not in globals():
    series_dict = {
        "Germany": df_wide["Germany"].dropna().astype(float),
        "Turkiye": df_wide["Turkiye"].dropna().astype(float)
    }

if "country_colors" not in globals():
    country_colors = {
        "Germany": "#4C78A8",
        "Turkiye": "#C46A3A"
    }

if "grid_color" not in globals():
    grid_color = "#D9E0E8"

if "spine_color" not in globals():
    spine_color = "#C7D0DA"

if "text_color" not in globals():
    text_color = "#27313F"

if "style_report_axes" not in globals():
    def style_report_axes(ax):
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        ax.spines["left"].set_color(spine_color)
        ax.spines["bottom"].set_color(spine_color)
        ax.spines["left"].set_linewidth(0.8)
        ax.spines["bottom"].set_linewidth(0.8)
        ax.grid(axis="y", color=grid_color, linewidth=0.8, alpha=0.85)
        ax.grid(axis="x", color=grid_color, linewidth=0.6, alpha=0.4)
        ax.set_axisbelow(True)
        ax.tick_params(axis="both", labelsize=11, length=0, pad=6)
        ax.margins(x=0.015)

def format_p_value_section3(p_value):
    if p_value < 0.001:
        return "< 0.001"
    return f"{p_value:.3f}"

def scan_chow_breaks(series, min_segment=8):
    y = series.dropna().astype(float)
    years = y.index.astype(int).to_numpy()
    values = y.to_numpy()
    rows = []
    for break_year in years[min_segment - 1: -(min_segment)]:
        mask = years <= break_year
        X_full = sm.add_constant(years.astype(float))
        X_1 = sm.add_constant(years[mask].astype(float))
        X_2 = sm.add_constant(years[~mask].astype(float))
        model_full = sm.OLS(values, X_full).fit()
        model_1 = sm.OLS(values[mask], X_1).fit()
        model_2 = sm.OLS(values[~mask], X_2).fit()
        rss_full = float((model_full.resid ** 2).sum())
        rss_split = float((model_1.resid ** 2).sum() + (model_2.resid ** 2).sum())
        k = X_full.shape[1]
        n = len(values)
        chow_stat = ((rss_full - rss_split) / k) / (rss_split / (n - 2 * k))
        p_value = 1 - f.cdf(chow_stat, k, n - 2 * k)
        rows.append({
            "break_year": int(break_year),
            "statistic": float(chow_stat),
            "p_value": float(p_value)
        })
    return pd.DataFrame(rows).sort_values("statistic", ascending=False).reset_index(drop=True)

def chow_test_for_year(series, break_year):
    years = series.dropna().index.astype(int).to_numpy()
    values = series.dropna().astype(float).to_numpy()
    mask = years <= break_year
    X_full = sm.add_constant(years.astype(float))
    X_1 = sm.add_constant(years[mask].astype(float))
    X_2 = sm.add_constant(years[~mask].astype(float))
    model_full = sm.OLS(values, X_full).fit()
    model_1 = sm.OLS(values[mask], X_1).fit()
    model_2 = sm.OLS(values[~mask], X_2).fit()
    rss_full = float((model_full.resid ** 2).sum())
    rss_split = float((model_1.resid ** 2).sum() + (model_2.resid ** 2).sum())
    k = X_full.shape[1]
    n = len(values)
    chow_stat = ((rss_full - rss_split) / k) / (rss_split / (n - 2 * k))
    p_value = 1 - f.cdf(chow_stat, k, n - 2 * k)
    return chow_stat, p_value

def build_trend_test_table(series_map):
    rows = []
    for country, series in series_map.items():
        runs_stat, runs_p = runstest_1samp(series, cutoff="median", correction=False)
        rows.append({
            "country": country,
            "test": "Runs test (median)",
            "наличие тренда": "Да" if runs_p < 0.05 else "Нет",
            "H0": "Последовательность случайна относительно медианы, систематического тренда нет.",
            "статистика": round(float(runs_stat), 4),
            "p-value": float(runs_p),
            "вывод": "H0 отвергается: порядок наблюдений неслучаен, тренд вероятен." if runs_p < 0.05 else "H0 не отвергается: явных признаков тренда тест не дал."
        })

        tau_stat, tau_p = kendalltau(series.index.astype(int), series.values)
        rows.append({
            "country": country,
            "test": "Kendall tau trend test",
            "наличие тренда": "Да" if tau_p < 0.05 else "Нет",
            "H0": "Монотонной связи между временем и уровнем ряда нет.",
            "статистика": round(float(tau_stat), 4),
            "p-value": float(tau_p),
            "вывод": "H0 отвергается: присутствует статистически значимый монотонный тренд." if tau_p < 0.05 else "H0 не отвергается: значимый монотонный тренд не подтвержден."
        })
    return pd.DataFrame(rows)

def build_trend_summary_table(trend_tests_table):
    rows = []
    for country, group in trend_tests_table.groupby("country"):
        runs_row = group[group["test"] == "Runs test (median)"].iloc[0]
        kendall_row = group[group["test"] == "Kendall tau trend test"].iloc[0]
        trend_present = runs_row["наличие тренда"] == "Да" and kendall_row["наличие тренда"] == "Да"
        rows.append({
            "country": country,
            "наличие тренда": "Да" if trend_present else "Нет",
            "H0": "Систематического тренда нет: ряд случаен относительно медианы и не имеет монотонной связи со временем.",
            "статистика": f"Runs z={runs_row['статистика']}; Kendall tau={kendall_row['статистика']}",
            "p-value": f"Runs {format_p_value_section3(float(runs_row['p-value']))}; Kendall {format_p_value_section3(float(kendall_row['p-value']))}",
            "вывод": "Оба теста отвергают H0, следовательно тренд статистически подтверждается." if trend_present else "Хотя бы один тест не отвергает H0, поэтому наличие тренда нельзя считать полностью подтвержденным."
        })
    return pd.DataFrame(rows)

def build_structural_break_table(series_map, break_years):
    rows = []
    for country, series in series_map.items():
        break_year = break_years[country]
        chow_stat, chow_p = chow_test_for_year(series, break_year)
        rows.append({
            "country": country,
            "наличие структурного сдвига": "Да" if chow_p < 0.05 else "Нет",
            "H0": f"В линейном тренде нет структурного сдвига в {break_year} году.",
            "статистика": round(float(chow_stat), 4),
            "p-value": format_p_value_section3(float(chow_p)),
            "вывод": "H0 отвергается: структурный сдвиг присутствует." if chow_p < 0.05 else "H0 не отвергается: убедительных признаков структурного сдвига нет."
        })
    return pd.DataFrame(rows)

trend_tests_table = build_trend_test_table(series_dict)
trend_summary_table = build_trend_summary_table(trend_tests_table)
break_scan_results = {country: scan_chow_breaks(series) for country, series in series_dict.items()}
candidate_break_years = {country: int(result.loc[0, "break_year"]) for country, result in break_scan_results.items()}
structural_break_table = build_structural_break_table(series_dict, candidate_break_years)
```

### 3.2. Проверка гипотезы об отсутствии тренда

Для проверки гипотезы об отсутствии тренда используем два теста, не связанные с единичным корнем.

- Теперь в разделе используются три теста: `ADF`, `KPSS` и `Phillips-Perron`.
- `Phillips-Perron` также проверяет нулевую гипотезу о наличии единичного корня, как и `ADF`, но отличается тем, что корректирует статистику непараметрически и потому менее чувствителен к автокорреляции и гетероскедастичности ошибок.
- `KPSS` полезен как парный тест с обратной нулевой гипотезой, а `PP` нужен как дополнительная проверка вывода `ADF` на коротком макроэкономическом ряде.
- Преимущество `ADF` в том, что он напрямую проверяет наличие единичного корня, но его слабая сторона — чувствительность к выбору лагов и относительно невысокая мощность на коротких выборках.
- Преимущество `KPSS` в том, что у него обратная нулевая гипотеза, поэтому он хорошо работает как парный тест к `ADF`; его недостаток в том, что результаты могут искажаться при трендах и структурных сдвигах.
- Преимущество `PP` в том, что он устойчивее к общей форме зависимости в ошибках без явного задания лаговой структуры, но на малых выборках тоже может давать нестабильные результаты.
- Основная проблема в наших рядах — возможные противоречия между `ADF`, `KPSS` и `PP`, особенно для Germany после первой разности. Это типичная ситуация для коротких макроэкономических рядов с трендом и вероятными структурными изменениями.

### 4.3. Оценка моделей

Ниже оцениваются по пять моделей для каждой страны: две обычные `ARIMA(p,d,q)` и три расширения с линейным трендом, параболическим трендом и dummy-переменной структурного сдвига. Анализ остатков и прогнозирование здесь специально не выполняются.

```python
import warnings
import numpy as np
from statsmodels.tsa.arima.model import ARIMA

warnings.filterwarnings("ignore")

break_years_section4 = globals().get("candidate_break_years", {"Germany": 2009, "Turkiye": 2003})

selected_model_specs = {
    "Germany": [
        {"model": "ARIMA(1,1,0)", "order": (1, 1, 0), "exog": "none"},
        {"model": "ARIMA(0,1,1)", "order": (0, 1, 1), "exog": "none"},
        {"model": "ARIMA(1,1,0)+linear trend", "order": (1, 1, 0), "exog": "linear"},
        {"model": "ARIMA(1,1,0)+quadratic trend", "order": (1, 1, 0), "exog": "quadratic"},
        {"model": "ARIMA(1,1,0)+dummy", "order": (1, 1, 0), "exog": "dummy"}
    ],
    "Turkiye": [
        {"model": "ARIMA(0,1,0)", "order": (0, 1, 0), "exog": "none"},
        {"model": "ARIMA(0,1,1)", "order": (0, 1, 1), "exog": "none"},
        {"model": "ARIMA(0,1,0)+linear trend", "order": (0, 1, 0), "exog": "linear"},
        {"model": "ARIMA(0,1,0)+quadratic trend", "order": (0, 1, 0), "exog": "quadratic"},
        {"model": "ARIMA(0,1,0)+dummy", "order": (0, 1, 0), "exog": "dummy"}
    ]
}

def build_section4_exog(kind, index, break_year):
    t = np.arange(1, len(index) + 1, dtype=float)
    years = index.astype(int)
    if kind == "none":
        return None
    if kind == "linear":
        return pd.DataFrame({"trend_linear": t}, index=index)
    if kind == "quadratic":
        return pd.DataFrame({"trend_linear": t, "trend_quadratic": t ** 2}, index=index)
    if kind == "dummy":
        return pd.DataFrame({"shift_dummy": (years >= break_year).astype(int)}, index=index)
    raise ValueError(kind)

def section4_root_status(result):
    if len(result.arroots):
        ar_status = "выполнена" if np.all(np.abs(result.arroots) > 1) else "не выполнена"
    else:
        ar_status = "неприменимо"
    if len(result.maroots):
        ma_status = "выполнена" if np.all(np.abs(result.maroots) > 1) else "не выполнена"
    else:
        ma_status = "неприменимо"
    return f"стационарность: {ar_status}; обратимость: {ma_status}"

def section4_pvalue(p_value):
    if p_value < 0.001:
        return "< 0.001"
    return f"{p_value:.3f}"

model_fit_rows = []
coefficient_rows = []

for country, specs in selected_model_specs.items():
    fitted_models = []
    for spec in specs:
        series = series_dict[country]
        exog = build_section4_exog(spec["exog"], series.index, break_years_section4[country])
        result = ARIMA(
            series,
            order=spec["order"],
            exog=exog,
            trend="n",
            enforce_stationarity=False,
            enforce_invertibility=False
        ).fit()
        fitted_models.append((spec, result))

        for parameter in result.params.index:
            p_value = float(result.pvalues[parameter])
            coefficient_rows.append({
                "country": country,
                "model": spec["model"],
                "parameter": parameter,
                "estimate": round(float(result.params[parameter]), 4),
                "p-value": section4_pvalue(p_value),
                "significance": "значим" if p_value < 0.05 else "не значим"
            })

    best_aic = min(result.aic for _, result in fitted_models)
    best_bic = min(result.bic for _, result in fitted_models)

    for spec, result in fitted_models:
        significant_parameters = [
            parameter for parameter in result.pvalues.index
            if parameter != "sigma2" and float(result.pvalues[parameter]) < 0.05
        ]

        if abs(result.aic - best_aic) < 1 or abs(result.bic - best_bic) < 1:
            note = "одна из сильнейших по AIC/BIC; окончательный выбор возможен только после диагностики остатков"
        elif significant_parameters:
            note = "содержательно полезная спецификация, но по информационным критериям уступает лидерам"
        else:
            note = "резервная модель: ключевые преимущества по AIC/BIC и значимости выражены слабо"

        model_fit_rows.append({
            "country": country,
            "модель": spec["model"],
            "параметры (p,d,q)": str(spec["order"]),
            "AIC": round(float(result.aic), 2),
            "BIC": round(float(result.bic), 2),
            "стационарность / обратимость": section4_root_status(result),
            "краткий вывод": note
        })

section4_coefficients = pd.DataFrame(coefficient_rows)
germany_model_coefficients = section4_coefficients[section4_coefficients["country"] == "Germany"].reset_index(drop=True)
turkiye_model_coefficients = section4_coefficients[section4_coefficients["country"] == "Turkiye"].reset_index(drop=True)
section4_model_comparison = pd.DataFrame(model_fit_rows)

display(section4_model_comparison)
```

Математические формы моделей для Germany:
- `ARIMA(1,1,0)`: `(1 - φ₁L)(1 - L)y_t = ε_t`
- `ARIMA(0,1,1)`: `(1 - L)y_t = (1 + θ₁L)ε_t`
- `ARIMA(1,1,0)+linear trend`: `(1 - φ₁L)(1 - L)y_t = β₁t + ε_t`
- `ARIMA(1,1,0)+quadratic trend`: `(1 - φ₁L)(1 - L)y_t = β₁t + β₂t² + ε_t`
- `ARIMA(1,1,0)+dummy`: `(1 - φ₁L)(1 - L)y_t = δD_t + ε_t`, где `D_t` — dummy структурного сдвига.

- `ARIMA(0,1,0)+linear trend`: `(1 - L)y_t = β₁t + ε_t`

### 4.4. Анализ остатков

Для всех моделей из подпункта 4.3 проверим остатки на автокорреляцию и нормальность распределения. Это позволит перейти от чисто информационных критериев к более содержательному выбору рабочей модели.

```python
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.stats.stattools import jarque_bera

def section4_exog_with_counter(kind, index, break_year, start_counter=1):
    t = np.arange(start_counter, start_counter + len(index), dtype=float)
    years = pd.Index(index).astype(int)
    if kind == "none":
        return None
    if kind == "linear":
        return pd.DataFrame({"trend_linear": t}, index=index)
    if kind == "quadratic":
        return pd.DataFrame({"trend_linear": t, "trend_quadratic": t ** 2}, index=index)
    if kind == "dummy":
        return pd.DataFrame({"shift_dummy": (years >= break_year).astype(int)}, index=index)
    raise ValueError(kind)

def build_future_section4_exog(kind, history_index, steps, break_year):
    future_index = pd.Index(range(int(pd.Index(history_index).astype(int).max()) + 1, int(pd.Index(history_index).astype(int).max()) + steps + 1))
    start_counter = len(history_index) + 1
    return future_index, section4_exog_with_counter(kind, future_index, break_year, start_counter=start_counter)

section4_diagnostic_rows = []
section4_model_results = {}

for country, specs in selected_model_specs.items():
    for spec in specs:
        key = (country, spec["model"])
        series = series_dict[country]
        exog = section4_exog_with_counter(spec["exog"], series.index, break_years_section4[country], start_counter=1)
        result = ARIMA(
            series,
            order=spec["order"],
            exog=exog,
            trend="n",
            enforce_stationarity=False,
            enforce_invertibility=False
        ).fit()
        residuals = pd.Series(result.resid).dropna()
        ljung_box = acorr_ljungbox(residuals, lags=[6], return_df=True)
        lb_stat = float(ljung_box["lb_stat"].iloc[0])
        lb_pvalue = float(ljung_box["lb_pvalue"].iloc[0])
        jb_stat, jb_pvalue, _, _ = jarque_bera(residuals)

        section4_model_results[key] = {
            "spec": spec,
            "result": result
        }

        section4_diagnostic_rows.append({
            "страна": country,
            "модель": spec["model"],
            "стационарность / обратимость": section4_root_status(result),
            "AIC": round(float(result.aic), 2),
            "BIC": round(float(result.bic), 2),
            "автокорреляция остатков (да/нет)": "Да" if lb_pvalue < 0.05 else "Нет",
            "нормальность (да/нет)": "Да" if jb_pvalue >= 0.05 else "Нет",
            "итоговый вывод": (
                "Остатки не показывают автокорреляцию, но нормальность отвергается."
                if lb_pvalue >= 0.05 and jb_pvalue < 0.05
                else "Остатки удовлетворяют базовым диагностическим требованиям."
                if lb_pvalue >= 0.05 and jb_pvalue >= 0.05
                else "Есть проблемы с остатками, модель стоит считать менее надежной."
            )
        })

section4_residual_diagnostics = pd.DataFrame(section4_diagnostic_rows)
display(section4_residual_diagnostics)
```

- Для всех оцененных моделей существенной автокорреляции остатков не обнаружено по тесту `Ljung-Box`.
- При этом нормальность остатков по тесту `Jarque-Bera` отвергается во всех случаях, поэтому доверительные интервалы и выводы по прогнозу следует интерпретировать аккуратно.
- Выбор рабочей модели поэтому делаем по совокупности факторов: `AIC`, `BIC` и отсутствию автокорреляции остатков.

### 4.5. Прогноз

Построим прогноз на 1–6 шагов вперед для выбранной модели каждой страны: с точечными значениями и 95% доверительными интервалами.

```python
forecast_rows = []
selected_forecasts = {}
selected_fitted_values = {}

for country, model_name in [("Germany", "ARIMA(1,1,0)+quadratic trend"), ("Turkiye", "ARIMA(0,1,0)+linear trend")]:
    model_bundle = section4_model_results[(country, model_name)]
    spec = model_bundle["spec"]
    result = model_bundle["result"]
    series = series_dict[country]
    fitted = pd.Series(np.asarray(result.fittedvalues), index=series.index).astype(float)
    future_index, future_exog = build_future_section4_exog(spec["exog"], series.index, 6, break_years_section4[country])
    forecast = result.get_forecast(steps=6, exog=future_exog)
    forecast_mean = pd.Series(np.asarray(forecast.predicted_mean), index=future_index).astype(float)
    forecast_ci = forecast.conf_int()
    forecast_ci.index = future_index

    selected_fitted_values[country] = fitted
    selected_forecasts[country] = {
        "mean": forecast_mean,
        "ci": forecast_ci
    }

    for step, year in enumerate(future_index, start=1):
        lower_col = forecast_ci.columns[0]
        upper_col = forecast_ci.columns[1]
        forecast_rows.append({
            "страна": country,
            "шаг": step,
            "год": int(year),
            "точечный прогноз": round(float(forecast_mean.loc[year]), 4),
            "нижняя граница 95%": round(float(forecast_ci.loc[year, lower_col]), 4),
            "верхняя граница 95%": round(float(forecast_ci.loc[year, upper_col]), 4)
        })

forecast_table = pd.DataFrame(forecast_rows)
display(forecast_table)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14.2, 5.0), dpi=160)

for ax, country in zip(axes, ["Germany", "Turkiye"]):
    series = series_dict[country].copy()
    fitted = selected_fitted_values[country].copy()
    forecast_mean = selected_forecasts[country]["mean"]
    forecast_ci = selected_forecasts[country]["ci"]

    lower_col = forecast_ci.columns[0]
    upper_col = forecast_ci.columns[1]

    fitted = pd.Series(fitted, index=fitted.index).reindex(series.index).dropna()

    threshold = series.min() * 0.8
    fitted = fitted[fitted >= threshold]

    ax.plot(
        series.index,
        series.values,
        color=country_colors[country],
        linewidth=2.4,
        alpha=0.9,
        label="Исходный ряд",
        solid_capstyle="round"
    )

    ax.plot(
        fitted.index,
        fitted.values,
        color="#6E7781",
        linewidth=1.8,
        alpha=0.9,
        linestyle="--",
        label="Fitted values"
    )

    ax.plot(
        forecast_mean.index,
        forecast_mean.values,
        color="#1F3B5B",
        linewidth=2.6,
        alpha=0.95,
        label="Forecast",
        solid_capstyle="round"
    )

    ax.fill_between(
        forecast_ci.index.astype(float),
        forecast_ci[lower_col].astype(float),
        forecast_ci[upper_col].astype(float),
        color="#9FB7CF",
        alpha=0.28,
        label="95% CI"
    )

    ax.set_title(f"Прогноз по выбранной модели: {country}", pad=12, color=text_color, fontweight="semibold")
    ax.set_xlabel("Год")
    ax.set_ylabel("% от общей занятости")
    style_report_axes(ax)
    ax.legend(loc="upper left", frameon=False, handlelength=2.8)

fig.tight_layout()
plt.show()
```

- В прогнозе для Germany показатель остается высоким и меняется очень плавно: прогнозная траектория близка к стабилизации около текущего уровня.
- Для Turkiye прогноз продолжает восходящее движение заметно быстрее, чем у Germany.
- Следовательно, в краткосрочном горизонте рост у Turkiye выглядит более быстрым, а у Germany более стабильным и инерционным.

### 4.6. Качество прогноза

Оценим качество выбранных моделей по стандартным in-sample метрикам на fitted values: `MAE`, `RMSE` и `MAPE`.

```python
quality_rows = []

for country, model_name in [("Germany", "ARIMA(1,1,0)+quadratic trend"), ("Turkiye", "ARIMA(0,1,0)+linear trend")]:
    actual = series_dict[country]
    fitted = selected_fitted_values[country].reindex(actual.index)
    mae = float(np.mean(np.abs(actual - fitted)))
    rmse = float(np.sqrt(np.mean((actual - fitted) ** 2)))
    mape = float(np.mean(np.abs((actual - fitted) / actual)) * 100)

    quality_rows.append({
        "страна": country,
        "MAE": round(mae, 4),
        "RMSE": round(rmse, 4),
        "MAPE": round(mape, 4),
        "вывод": (
            "Прогноз более точен по относительной ошибке." if country == "Germany" else
            "Прогноз менее точен по MAPE, но сохраняет приемлемую ошибку на фоне более быстрого роста ряда."
        )
    })

forecast_quality_table = pd.DataFrame(quality_rows)
display(forecast_quality_table)
```

- По `MAPE` прогноз точнее у Germany, то есть в относительном выражении модель для Germany ошибается меньше.
- По `RMSE` и `MAE` значения нужно интерпретировать вместе с масштабом ряда, но в целом обе модели дают приемлемую краткосрочную аппроксимацию.
- Надежность прогноза можно считать умеренной: отсутствие автокорреляции остатков поддерживает рабочее качество моделей, но ненормальность остатков требует осторожности при интерпретации интервалов.

- Для Germany выбрана модель `ARIMA(1,1,0)+quadratic trend`, для Turkiye — `ARIMA(0,1,0)+linear trend`.
- Germany и Turkiye различаются тем, что германский ряд лучше описывается более плавной и гибкой трендовой траекторией, а турецкий — более простой моделью роста с выраженным восходящим движением.
- Прогноз для Germany выглядит более стабильным и осторожным, а для Turkiye — более быстрым по росту. При этом в обоих случаях надежность прогноза ограничена ненормальностью остатков, поэтому результаты следует трактовать как рабочий, а не окончательно безусловный ориентир.
