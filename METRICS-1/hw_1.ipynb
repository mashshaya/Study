Задача 6
Найди коэффициент при X2 в регрессии Y на X1 и X2

```python
import statsmodels.api as sm

X = df[["X1", "X2"]]  
X = sm.add_constant(X)  
y = df["Y"]

model = sm.OLS(y, X).fit()
round(model.params["X2"], 3)
```

Задача 7
Найди R^2 в регрессии Y на X1 и X2

```python
round(model.rsquared, 3)
```

Задача 8
Найди t-статистику для гипотезы о равенстве константы нулю в
регрессии Y на X1 и X2

```python
round(model.tvalues["const"], 3)
```

Задача 9
Найди оценку дисперсии ошибок в регрессии Y на X1 и X2

```python
round(model.mse_resid, 3)
```
