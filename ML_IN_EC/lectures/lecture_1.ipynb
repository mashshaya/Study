# Лекция 1. Байесовский классификатор, наивный Байес и байесовские сети

---

# 1. Постановка задачи классификации

Есть:

- целевая переменная  
$$
Y \in \{0,1\}
$$

- признаки  
$$
X = (X_1, ..., X_m)
$$

Задача — оценить:

$$
P(Y=1 \mid X=x)
$$

и построить правило:

$$
\hat y(x) = 
\begin{cases}
1, & \text{если } P(Y=1 \mid X=x) \ge c \\
0, & \text{иначе}
\end{cases}
$$

Обычно:
$$
c = 0.5
$$

**Интуиция:** выбираем класс с большей вероятностью.

---

# 2. Оценивание безусловных вероятностей

$$
\hat p = \frac{1}{n} \sum_{i=1}^n Y_i
$$

**Как это работает:**
- суммируем все единицы
- делим на число наблюдений

**Мини-пример:**  
\( Y = [1,0,1,1] \Rightarrow \hat p = 3/4 = 0.75 \)

**Интуиция:** вероятность = частота события.

---

# 3. Условная вероятность

$$
P(Y=1 \mid X=x) = \frac{P(Y=1, X=x)}{P(X=x)}
$$

Оценка:

$$
\hat p_x = \frac{n(x,y)}{n(x)}
$$

где:
- \( n(x) \) — сколько раз встретилось \( X=x \)
- \( n(x,y) \) — сколько из них имеют \( Y=1 \)

**Мини-пример:**
если среди 10 наблюдений с \( X=x \) в 7 случаях \( Y=1 \):

$$
\hat p_x = 7/10 = 0.7
$$

**Интуиция:** доля единиц внутри группы.

---

# 4. Индикаторная запись

Индикатор:

$$
I(A) =
\begin{cases}
1, & \text{если } A \text{ верно} \\
0, & \text{иначе}
\end{cases}
$$

Тогда:

$$
\hat p_x =
\frac{\sum_{i=1}^n I(X_i = x, Y_i = 1)}
{\sum_{i=1}^n I(X_i = x)}
$$

**Как это работает:**
- числитель считает нужные строки
- знаменатель считает все строки с \( X=x \)

---

# 5. Байесовский классификатор

$$
\hat y(x) = I(\hat p_x \ge c)
$$

**Как это работает:**
1. считаем вероятность
2. сравниваем с порогом
3. выдаем класс

---

# 6. Проклятие размерности

Проблема:

- число комбинаций \( X \) растет как:
$$
2^m
$$

- для многих комбинаций:
$$
n(x) \approx 0
$$

Следствия:
- оценки шумные (высокая дисперсия)
- иногда невозможно посчитать вероятность

**Интуиция:** данных не хватает на все комбинации признаков.

---

# 7. Уменьшение размерности

Идея:

$$
P(Y \mid X_1, X_2, X_3) \approx P(Y \mid X_1, X_2)
$$

**Что происходит:**
- игнорируем слабый признак
- увеличиваем размер выборки внутри группы

**Интуиция:** лучше чуть ошибиться (смещение), чем иметь сильный шум (дисперсию).

---

# 8. Независимость

## Обычная независимость

$$
P(X_1, X_2) = P(X_1)P(X_2)
$$

## Условная независимость

$$
P(X_1, X_2 \mid Y) = P(X_1 \mid Y)P(X_2 \mid Y)
$$

**Интуиция:**
- переменные могут быть зависимы в целом
- но независимы при фиксированном \( Y \)

---

# 9. Формула Байеса

$$
P(Y \mid X) =
\frac{P(Y)P(X \mid Y)}{P(X)}
$$

**Как это работает:**
- \( P(Y) \) — априорная вероятность
- \( P(X \mid Y) \) — насколько признаки типичны для класса
- \( P(X) \) — нормировка

**Интуиция:** обновляем вероятность класса после наблюдения признаков.

---

# 10. Наивный Байес

Главное допущение:

$$
P(X \mid Y) = \prod_{j=1}^m P(X_j \mid Y)
$$

Подставляем в Байеса:

$$
P(Y \mid X) =
\frac{P(Y)\prod_{j=1}^m P(X_j \mid Y)}
{P(X)}
$$

Для сравнения классов:

$$
P(Y=1 \mid X) \propto P(Y=1)\prod_{j=1}^m P(X_j \mid Y=1)
$$

---

# 11. Как считается на практике

## Априорная вероятность

$$
\hat P(Y=1) = \frac{\sum Y_i}{n}
$$

## Условные вероятности

$$
\hat P(X_j \mid Y=1) =
\frac{\text{число наблюдений с } X_j \text{ и } Y=1}
{\text{число наблюдений с } Y=1}
$$

---

# 12. Полная вероятность

$$
P(X) = \sum_{y \in \{0,1\}} P(X \mid Y=y)P(Y=y)
$$

**Интуиция:** учитываем оба класса.

---

# 13. Почему работает наивный Байес

Без упрощения нужно оценивать:

$$
P(X_1, ..., X_m \mid Y)
$$

Это требует много данных.

С упрощением:

$$
\prod_{j=1}^m P(X_j \mid Y)
$$

**Интуиция:** заменяем сложную вероятность на произведение простых → резко уменьшаем требования к данным.

---

# 14. Смещение vs дисперсия

Наивный Байес:

- увеличивает смещение (ошибка в модели)
- сильно снижает дисперсию (шум)

$$
\text{bias} \uparrow, \quad \text{variance} \downarrow
$$

---

# 15. Байесовские сети

Модель — ориентированный ациклический граф (DAG).

Факторизация:

$$
P(X_1, ..., X_m) = \prod_{j=1}^m P(X_j \mid \text{Parents}(X_j))
$$

**Как это работает:**
- каждая переменная зависит только от родителей
- не нужно хранить полное совместное распределение

---

# 16. Частные случаи

## Наивный Байес — это сеть:

$$
Y \rightarrow X_1, ..., X_m
$$

## Общая сеть — произвольная структура зависимостей

---

# Итог

Основная цель:

$$
P(Y \mid X)
$$

Подходы:

1. Байесовский классификатор — точный, но требует много данных  
2. Наивный Байес — упрощение через независимость  
3. Байесовские сети — гибкая факторизация распределения  

Ключевая идея всей лекции:

$$
\text{упрощать вероятности так, чтобы их можно было оценить по данным}
$$
