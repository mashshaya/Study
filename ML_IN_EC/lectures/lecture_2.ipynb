# Лекция 2. Метод ближайших соседей

---

# 1. Основная идея метода

Метод ближайших соседей, или kNN, строит прогноз для нового объекта по тем объектам из обучающей выборки, которые находятся к нему ближе всего по признакам.

Идея метода очень простая:

- похожие объекты должны иметь похожие значения целевой переменной;
- сначала нужно определить, что значит "похожи";
- затем нужно посмотреть на ближайших наблюдений и использовать их целевые значения для прогноза.

Главная особенность метода: он почти не делает структурных предположений о форме зависимости между признаками и целевой переменной. Поэтому метод гибкий, но сильно зависит от того, как именно мы измеряем расстояние между объектами. 

---

# 2. Метрика расстояния

Чтобы сравнивать объекты, вводится функция расстояния между двумя векторами признаков.

Пусть есть два объекта:

$$
x = (x_1, \dots, x_m), \qquad y = (y_1, \dots, y_m)
$$

Тогда расстояние между ними задается функцией

$$
d(x, y)
$$

Чтобы функция была метрикой, она должна удовлетворять свойствам:

1. неотрицательность:
$$
d(x,y) \ge 0
$$

2. тождественность:
$$
d(x,y) = 0 \iff x = y
$$

3. симметричность:
$$
d(x,y) = d(y,x)
$$

4. неравенство треугольника:
$$
d(x,y) \le d(x,z) + d(z,y)
$$

## Интуиция

Метрика формализует понятие "сходства". Чем меньше расстояние, тем объекты ближе друг к другу по признакам.

---

# 3. Расстояние Минковского

Самая общая популярная форма расстояния в лекции — расстояние Минковского:

$$
d(x,y)=\left(\sum_{i=1}^{m} |x_i-y_i|^{\lambda}\right)^{1/\lambda}
$$

где

$$
\lambda \in \mathbb{N}
$$

## Как работает эта формула

1. Для каждого признака берется разница:
$$
|x_i-y_i|
$$

2. Каждая разница возводится в степень:
$$
|x_i-y_i|^\lambda
$$

3. Все такие вклады суммируются.

4. Затем берется степень:
$$
1/\lambda
$$

## Интуиция

Чем больше параметр $$\lambda$$, тем сильнее штрафуются крупные расхождения по отдельным признакам.

---

# 4. Евклидово расстояние

При

$$
\lambda = 2
$$

получаем евклидово расстояние:

$$
d(x,y)=\sqrt{\sum_{i=1}^{m}(x_i-y_i)^2}
$$

## Как работает

- измеряет "обычную" длину отрезка между двумя точками в пространстве признаков;
- крупные различия штрафуются сильнее, потому что они возводятся в квадрат.

## Мини-пример

Если

$$
x=(1,2), \qquad y=(4,6)
$$

то

$$
d(x,y)=\sqrt{(1-4)^2+(2-6)^2}
$$

$$
d(x,y)=\sqrt{9+16}=\sqrt{25}=5
$$

---

# 5. Манхэттенское расстояние

При

$$
\lambda = 1
$$

получаем манхэттенское расстояние:

$$
d(x,y)=\sum_{i=1}^{m}|x_i-y_i|
$$

## Как работает

- по каждому признаку берется модуль разницы;
- затем все различия просто складываются.

## Интуиция

Это расстояние считает, сколько нужно пройти "по координатам", а не по прямой. Оно менее чувствительно к одиночным большим отклонениям, чем евклидово.

## Мини-пример

Если

$$
x=(1,2), \qquad y=(4,6)
$$

то

$$
d(x,y)=|1-4|+|2-6|=3+4=7
$$

---

# 6. kNN с одним соседом

Если используется только один ближайший сосед, то прогноз для объекта $$x$$ равен целевой переменной ближайшего к нему объекта обучающей выборки.

Пусть обучающие наблюдения имеют признаки

$$
X_1, X_2, \dots, X_n
$$

и целевые значения

$$
Y_1, Y_2, \dots, Y_n
$$

Тогда правило классификации:

$$
\hat y(x)=Y_i,
\qquad
i=\arg\min_{j \in \{1,\dots,n\}} d(x,X_j)
$$

## Как работает формула

- для нового объекта $$x$$ считаем расстояния до всех объектов обучающей выборки;
- выбираем индекс объекта с минимальным расстоянием;
- берем его класс.

## Интуиция

Новый объект "копирует" ответ самого похожего наблюдения.

## Недостаток

Такой прогноз имеет высокую дисперсию: он полностью зависит от одного наблюдения.

---

# 7. kNN с несколькими соседями

Чтобы уменьшить нестабильность прогноза, берут не одного, а сразу $$k$$ ближайших соседей.

Тогда прогноз равен классу, который чаще встречается среди этих соседей.

Если среди ближайших соседей значения целевой переменной равны

$$
N_{y,1}, \dots, N_{y,k}
$$

то можно прогнозировать 1, если единиц большинство.

## Интуиция

Один сосед может быть случайным шумом, а несколько соседей дают более устойчивую локальную картину.

## Почему часто берут нечетное $$k$$

Чтобы реже возникала ничья между классами 0 и 1.

---

# 8. Спорные ближайшие соседи

Иногда несколько наблюдений находятся на одинаковом минимальном расстоянии от точки $$x$$.

Это означает, что ближайший сосед определен неоднозначно.

В лекции указано правило: если не сказано иное, при равенстве расстояний выбирается сосед с **наименьшим индексом**.

## Интуиция

Это просто техническое правило, чтобы алгоритм был однозначным и воспроизводимым.

---

# 9. Оценка условной вероятности в kNN

В kNN можно не только выдать класс, но и оценить условную вероятность класса 1.

Если среди $$k$$ ближайших соседей ровно $$N_Y$$ наблюдений имеют класс 1, то

$$
\hat P(Y_i=1 \mid X_i=x)=\frac{1}{k}\sum_{i=1}^{k} N_{y,i}=\frac{N_Y}{k}
$$

## Как работает формула

- считаем, сколько единиц среди соседей;
- делим на число соседей.

## Мини-пример

Если среди 5 ближайших соседей классы такие:

$$
1,\ 0,\ 1,\ 1,\ 0
$$

то единиц 3, поэтому

$$
\hat P(Y=1 \mid X=x)=\frac{3}{5}=0.6
$$

---

# 10. Порог классификации

После оценки вероятности нужно перевести ее в итоговый класс.

Используется правило:

$$
\hat y(x)=I(\hat p_x \ge c)
$$

где

- $$\hat p_x$$ — оценка условной вероятности;
- $$c$$ — порог;
- $$I(\cdot)$$ — индикатор.

Индикатор можно записать так:

$$
I(A)=
\begin{cases}
1, & \text{если условие } A \text{ верно} \\
0, & \text{иначе}
\end{cases}
$$

## Интуиция

Порог задает, насколько большой должна быть вероятность, чтобы мы решились предсказать класс 1.

---

# 11. Взвешенный kNN

Обычный kNN учитывает соседей одинаково. Но интуитивно ближний сосед должен быть важнее дальнего.

Поэтому вводятся веса, зависящие от расстояния. Например, весом может быть величина

$$
\frac{1}{d(N(x,i),x)}
$$

где $$N(x,i)$$ — признаки i-го ближайшего соседа.

Тогда классификатор можно записать так:

$$
\hat y(x)=I\left(
\sum_{i: N(y,i)=1}\frac{1}{d(N(x,i),x)}
\ge
\sum_{i: N(y,i)=0}\frac{1}{d(N(x,i),x)}
\right)
$$

Оценка условной вероятности:

$$
\hat P(Y_i=1 \mid X_i=x)=
\frac{
\sum_{i: N(y,i)=1}\frac{1}{d(N(x,i),x)}
}{
\sum_{i=1}^{k}\frac{1}{d(N(x,i),x)}
}
$$

## Как работает

- ближние соседи получают больший вес;
- дальние — меньший;
- вероятность становится взвешенной долей единиц.

## Интуиция

Если два соседа имеют класс 1, но один почти совпадает с точкой $$x$$, а другой далеко, то логично доверять первому сильнее.

---

# 12. Смещение и дисперсия в kNN

Для kNN есть классический компромисс:

- маленькое $$k$$:
  - низкое смещение;
  - высокая дисперсия.

- большое $$k$$:
  - ниже дисперсия;
  - выше смещение.

## Почему так

Если $$k$$ маленькое, мы очень точно подстраиваемся под локальную структуру данных, но сильно зависим от случайности.

Если $$k$$ большое, мы усредняем по большой окрестности, но можем замазать важные локальные различия.

---

# 13. Нормализация признаков

kNN очень чувствителен к масштабу признаков.

Если один признак измеряется в больших числах, а другой — в маленьких, первый может почти полностью определять расстояние.

## Стандартизация

Один из способов нормализации:

$$
x_j^{\text{norm}}=\frac{x_j-\bar x_j}{s_j}
$$

где

- $$\bar x_j$$ — выборочное среднее признака;
- $$s_j$$ — выборочное стандартное отклонение.

## Нормализация в отрезок $$[0,1]$$

Альтернативный способ:

$$
x_j^{\text{norm}}=\frac{x_j-\min(x_j)}{\max(x_j)-\min(x_j)}
$$

## Интуиция

После нормализации признаки становятся сопоставимыми по масштабу, и расстояние начинает отражать содержательное различие, а не просто разницу единиц измерения.

---

# 14. Индексы расстояния для категориальных и бинарных признаков

Обычные метрики расстояния хорошо работают для непрерывных признаков, но не всегда удобны для бинарных.

Если все признаки бинарные, можно использовать специальные индексы сходства.

## 14.1. Индекс Рассела–Рао

Это доля признаков, где оба объекта одновременно равны 1.

Если у двух объектов есть бинарные векторы, то индекс Рассела–Рао можно мыслить как

$$
\text{RR}(x,y)=\frac{\#\{j: x_j=1,\ y_j=1\}}{m}
$$

## Интуиция

Смотрим, насколько часто два объекта совпадают именно по "положительным" признакам.

---

## 14.2. Индекс Сокаля–Минхера

Это доля совпадающих значений вообще:

$$
\text{SM}(x,y)=\frac{\#\{j: x_j=y_j\}}{m}
$$

## Интуиция

Одинаково учитываются совпадения и по нулям, и по единицам.

---

## 14.3. Индекс Жаккара

В разреженных данных нулей очень много, поэтому совпадения по нулям часто неинформативны.

Тогда удобно использовать индекс Жаккара:

$$
\text{J}(x,y)=
\frac{\#\{j: x_j=1,\ y_j=1\}}
{\#\{j: x_j=1 \text{ или } y_j=1\}}
$$

## Интуиция

Мы сравниваем объекты только по активным признакам, игнорируя массовые совпадения по нулям.

---

# 15. Приблизительный поиск ближайших соседей

Если наблюдений очень много, то точный поиск ближайших соседей становится дорогим вычислительно.

Потому что для нового объекта приходится считать расстояния до всех объектов обучающей выборки.

Тогда используют приближенные методы поиска, например k-d tree.

## Интуиция

Иногда выгоднее найти не идеально ближайших соседей, а "почти ближайших", но гораздо быстрее.

---

# 16. Типы прогнозов в бинарной классификации

После классификации каждый прогноз относится к одному из четырех типов.

## Истинно положительный прогноз

TP — true positive:

- истина: 1
- прогноз: 1

## Истинно отрицательный прогноз

TN — true negative:

- истина: 0
- прогноз: 0

## Ложно положительный прогноз

FP — false positive:

- истина: 0
- прогноз: 1

## Ложно отрицательный прогноз

FN — false negative:

- истина: 1
- прогноз: 0

## Интуиция

Это полный словарь ошибок и верных решений модели.

---

# 17. Матрица ошибок

Количество прогнозов каждого типа удобно представлять в виде матрицы ошибок:

|                | Прогноз = 1 | Прогноз = 0 |
|----------------|-------------|-------------|
| Истина = 1     | TP          | FN          |
| Истина = 0     | FP          | TN          |

Эта таблица — база почти для всех метрик качества в бинарной классификации.

---

# 18. Accuracy

Точность классификации:

$$
ACC=\frac{TP+TN}{TP+TN+FP+FN}
$$

## Как работает

- в числителе число верных прогнозов;
- в знаменателе число всех прогнозов.

## Интуиция

Доля объектов, которые модель классифицировала правильно.

## Ограничение

Если классы сильно несбалансированы, accuracy может быть обманчивой.

---

# 19. Как порог влияет на TP, TN, FP, FN

Если увеличить порог $$c$$ в правиле

$$
\hat y(x)=I(\hat p_x \ge c)
$$

то единицы будут предсказываться реже.

Из-за этого обычно:

- $$TN$$ растет;
- $$FN$$ растет;
- $$TP$$ падает;
- $$FP$$ падает.

## Интуиция

Высокий порог делает модель "осторожной": она реже ставит класс 1.

---

# 20. Подбор оптимального порога

Если разные типы прогнозов имеют разную цену, порог лучше подбирать не механически, а через целевую функцию.

Пусть:

- $$p_{TP}$$ — цена TP;
- $$p_{TN}$$ — цена TN;
- $$p_{FP}$$ — цена FP;
- $$p_{FN}$$ — цена FN.

Тогда оптимальный порог можно искать как решение задачи:

$$
c=
\arg\max_{c_i}
\left[
p_{TP}TP(c_i)+p_{TN}TN(c_i)+p_{FP}FP(c_i)+p_{FN}FN(c_i)
\right]
$$

## Как работает

1. сортируем оцененные вероятности;
2. по очереди берем каждую как кандидат на порог;
3. считаем итоговую выгоду;
4. выбираем лучший порог.

## Интуиция

Порог должен зависеть не только от статистики, но и от экономического смысла ошибок.

---

# 21. Precision

Точность положительных прогнозов:

$$
precision=\frac{TP}{TP+FP}
$$

## Как работает

Из всех объектов, которым модель поставила 1, считаем долю действительно положительных.

## Интуиция

Если модель сказала "это 1", насколько часто она права?

---

# 22. Recall

Полнота:

$$
recall=\frac{TP}{TP+FN}
$$

## Как работает

Из всех настоящих единиц считаем долю тех, кого модель нашла.

## Интуиция

Насколько хорошо модель ловит положительный класс?

---

# 23. F1-мера

F1 — гармоническое среднее precision и recall:

$$
F1=
\frac{2 \cdot precision \cdot recall}{precision+recall}
$$

## Почему именно гармоническое среднее

Оно сильно штрафует ситуацию, когда одна из двух метрик очень маленькая.

## Интуиция

F1 велика только тогда, когда и precision, и recall одновременно хорошие.

---

# 24. Несбалансированная выборка

Выборка называется несбалансированной, если объектов одного класса намного больше, чем другого.

Например, дефолтов обычно сильно меньше, чем недефолтов.

В такой ситуации обычная accuracy часто малоинформативна.

---

# 25. Recall для класса 0

Для отрицательного класса полнота считается так:

$$
recall(0)=\frac{TN}{TN+FP}
$$

Для положительного класса:

$$
recall(1)=\frac{TP}{TP+FN}
$$

---

# 26. Средняя полнота по классам

Один вариант:

$$
ACA=\frac{recall(1)+recall(0)}{2}
$$

Другой вариант — гармоническое среднее:

$$
ACA=
\frac{2 \cdot recall(1)\cdot recall(0)}
{recall(1)+recall(0)}
$$

## Интуиция

Эти метрики специально смотрят, как модель работает по обоим классам, а не только в среднем по выборке.

---

# 27. ROC-кривая

Для каждого порога можно посчитать:

## True Positive Rate

$$
TPR=\frac{TP}{TP+FN}
$$

Это то же самое, что recall для класса 1.

## False Positive Rate

$$
FPR=\frac{FP}{TN+FP}
$$

Это доля ложных тревог среди настоящих нулей.

Тогда ROC-кривая — это график зависимости

$$
TPR \text{ от } FPR
$$

при разных порогах.

## Интуиция

ROC показывает, как меняется способность модели ловить единицы ценой ложных срабатываний.

---

# 28. Интерпретация ROC

Если ROC-кривая одной модели лежит выше другой, то при том же уровне ложноположительных срабатываний она дает больший TPR.

Значит, в соответствующем диапазоне порогов она лучше.

---

# 29. AUC

AUC — это площадь под ROC-кривой:

$$
AUC = \int ROC
$$

В практическом смысле — агрегированная мера качества по всем порогам.

## Интерпретация

- $$AUC=1$$ — идеальная модель;
- $$AUC=0.5$$ — случайное угадывание;
- чем выше AUC, тем лучше в среднем модель разделяет классы.

## Ограничение

AUC усредняет качество по всем порогам. Иногда это плохо, потому что нас интересует только ограниченный диапазон порогов.

---

# 30. Частичный AUC

Если важен только определенный участок ROC-кривой, используют частичный AUC:

$$
pAUC(a,b)
$$

Это площадь под ROC-кривой на участке от $$a$$ до $$b$$ по оси FPR.

## Интуиция

Можно оценивать модель только там, где уровень ложноположительных ошибок приемлем в содержательной задаче.

---

# 31. Гиперпараметры kNN

Гиперпараметры — это параметры, которые не оцениваются из данных автоматически внутри метода, а задаются извне.

Для kNN ключевые гиперпараметры:

- число соседей $$k$$;
- метрика расстояния;
- способ нормализации;
- порог классификации;
- вариант взвешивания соседей.

---

# 32. Тюнинг гиперпараметров через кросс-валидацию

Общий алгоритм:

1. выбираем метрику качества, например ACC или F1;
2. перебираем разные комбинации гиперпараметров;
3. оцениваем модель по кросс-валидации;
4. выбираем комбинацию, максимизирующую метрику.

## Интуиция

Мы подбираем не сами параметры модели, а настройки метода.

---

# 33. Жадкий перебор

Если есть конечный список вариантов, можно перебрать все комбинации.

Например:

- $$k \in \{1,2,3\}$$
- метрика: манхэттенская или евклидова

Тогда просто строим модель для каждой пары и сравниваем качество.

## Недостаток

Если гиперпараметров много, полный перебор становится слишком дорогим.

---

# 34. Рандомизированный поиск

Если пространство гиперпараметров большое, можно перебирать их частично случайно.

## Интуиция

Не обязательно смотреть все варианты. Иногда достаточно проверить случайную подвыборку разумных комбинаций.

---

# 35. Валидационная выборка

Данные обычно делят на три части:

- обучающая выборка;
- валидационная выборка;
- тестовая выборка.

## Роли выборок

### Обучающая
Используется для обучения модели.

### Валидационная
Используется для подбора гиперпараметров.

### Тестовая
Используется только для финальной проверки качества.

## Почему нельзя тюнить на тесте

Если подгонять гиперпараметры под тестовую выборку, то тест перестает быть честной проверкой.

---

# 36. Gain

Пусть модель оценивает вероятности покупки, и мы сортируем клиентов по этим вероятностям по убыванию.

Тогда gain показывает, какую долю настоящих единиц мы поймали в верхней части списка.

Например, gain для первой децили:

$$
gain(1)=
\frac{\text{число единиц в 1-й децили}}
{\text{общее число единиц}}
$$

## Интуиция

Если мы берем только самых перспективных клиентов, какую долю всех потенциальных покупателей мы этим покрываем?

---

# 37. Кумулятивный gain

Для первых $$k$$ децилей:

$$
cumulative\ gain(k)=
\frac{\text{число единиц в первых } k \text{ децилях}}
{\text{общее число единиц}}
$$

## Интуиция

Показывает, насколько хорошо модель концентрирует единицы наверху ранжированного списка.

---

# 38. Lift

Lift сравнивает модель со случайным отбором.

Для k-й децили:

$$
lift(k)=
\frac{
\text{число единиц в } k\text{-й децили}
}{
\text{ожидаемое число единиц в } k\text{-й децили при случайном отборе}
}
$$

## Интерпретация

- $$lift>1$$ — модель лучше случайного выбора;
- $$lift=1$$ — не лучше случайности;
- $$lift<1$$ — хуже случайного выбора.

---

# 39. Кумулятивный lift

Для первых $$k$$ децилей:

$$
cumulative\ lift(k)=
\frac{
\text{число единиц в первых } k \text{ децилях}
}{
\text{ожидаемое число единиц в первых } k \text{ децилях при случайном отборе}
}
$$

## Интуиция

Показывает, во сколько раз использование модели лучше случайного таргетинга, если брать верхнюю часть ранжированного списка.

---

# 40. Непрерывные gain и lift

Вместо децилей можно рассматривать любую долю выборки.

Например, не 10%, 20%, 30%, а 57% лучших наблюдений.

Тогда строятся непрерывные кривые кумулятивного gain и cumulative lift.

## Интуиция

Это дает более детальное понимание, где именно использование модели еще выгодно, а где уже нет.

---

# 41. Нерепрезентативная выборка

Иногда несбалансированность вызвана не только самой природой данных, но и способом их сбора.

Тогда модель может завышать или занижать вероятности просто потому, что в выборке классы представлены не так, как в генеральной совокупности.

---

# 42. Undersampling

Undersampling — случайное удаление части объектов из избыточного класса.

## Интуиция

Если нулей слишком много, можно убрать часть нулей, чтобы сделать выборку более сбалансированной.

## Плюс

Упрощает задачу.

## Минус

Теряется информация.

---

# 43. Oversampling

Oversampling — дублирование объектов редкого класса.

## Интуиция

Если единиц слишком мало, можно искусственно увеличить их число в выборке.

## Плюс

Редкий класс начинает сильнее влиять на обучение.

## Минус

Есть риск переобучения, потому что одни и те же наблюдения повторяются.

---

# 44. Общий вывод по методу ближайших соседей

Метод ближайших соседей — это очень простой и гибкий непараметрический метод, который:

- строит прогноз по локальной структуре данных;
- не требует задания явной формы зависимости;
- чувствителен к выбору расстояния, масштабу признаков и числу соседей.

Его сильные стороны:

- простота;
- гибкость;
- понятная интерпретация;
- возможность сразу получать оценку условной вероятности.

Его слабые стороны:

- высокая чувствительность к масштабу признаков;
- вычислительная дороговизна на больших выборках;
- сильная зависимость от гиперпараметров;
- проблемы на несбалансированных и нерепрезентативных данных без дополнительной настройки.

Главная идея лекции:

$$
\text{хороший прогноз в kNN получается тогда, когда хорошо определено понятие близости}
$$
